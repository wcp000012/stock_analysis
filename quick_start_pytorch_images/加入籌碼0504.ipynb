```python
import twstock
import yfinance as yf
import pandas as pd
import numpy as np
import requests
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, precision_recall_curve
import matplotlib.pyplot as plt
import seaborn as sns
from concurrent.futures import ThreadPoolExecutor
import pickle
from torch.utils.data import TensorDataset, DataLoader
import gc
from imblearn.over_sampling import SMOTE
import os
from torch.cuda.amp import autocast, GradScaler

# 設置環境變數以優化內存分配
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# 獲取台灣股票列表
url = 'https://isin.twse.com.tw/isin/C_public.jsp?strMode=2'
response = requests.get(url)
response.encoding = 'big5'
dfs = pd.read_html(response.text)
stock_list = dfs[0]
stock_list.columns = stock_list.iloc[0]
stock_list = stock_list[1:]
stock_list = stock_list[stock_list['有價證券代號及名稱'].str.contains(r'\d{4}\s+')]
stock_list['股票代號'] = stock_list['有價證券代號及名稱'].str.split().str[0]
stock_list['股票名稱'] = stock_list['有價證券代號及名稱'].str.split().str[1]
stock_list = stock_list[['股票代號', '股票名稱']]
stock_codes = stock_list['股票代號'].tolist()

# 定義數據集生成函數（分類任務）
def create_dataset(data, time_step=126, future_days=120):
    X, Y = [], []
    for i in range(len(data) - time_step - future_days):
        X.append(data[i:(i + time_step), :])
        current_price = data[i + time_step, 0]
        future_price = data[i + time_step + future_days, 0]
        price_change = (future_price - current_price) / current_price
        Y.append(1 if price_change >= 0.5 else 0)  # 維持 50% 閾值
    return np.array(X), np.array(Y)

# 定義下載單支股票數據的函數
import twstock
import requests
import pandas as pd
import numpy as np

def download_stock_data(stock_code):
    try:
        stock_code_tw = stock_code + ".TW"
        start_date = '2015-01-01'
        end_date = '2024-12-31'
        
        # 下載價格數據 (yfinance)
        stock_data = yf.download(stock_code_tw, start=start_date, end=end_date, progress=False)
        if stock_data.empty or len(stock_data) < 120 + 126:
            return None, stock_code_tw

        # 計算技術指標
        stock_data['Volume'] = stock_data['Volume']
        stock_data['MA10'] = stock_data['Close'].rolling(window=10).mean()
        stock_data['RSI'] = compute_rsi(stock_data['Close'], 14)

        # 下載融資融券數據 (twstock)
        stock = twstock.Stock(stock_code)
        margin_data = stock.fetch_margin()
        margin_df = pd.DataFrame(margin_data)
        margin_df['date'] = pd.to_datetime(margin_df['date'])
        margin_df.set_index('date', inplace=True)
        margin_df = margin_df[['margin_buy', 'margin_sell', 'short_buy', 'short_sell']]  # 融資買賣、融券買賣

        # 下載三大法人買賣超數據 (TWSE CSV)
        twse_url = f"https://www.twse.com.tw/fund/T86?response=csv&date={end_date.replace('-','')}&selectType=ALLBUT0999"
        twse_data = pd.read_csv(twse_url, encoding='big5')
        twse_data = twse_data[twse_data['證券代號'] == stock_code]
        twse_data['日期'] = pd.to_datetime(twse_data['日期'], format='%Y%m%d')
        twse_data.set_index('日期', inplace=True)
        twse_data = twse_data[['外資買賣超', '投信買賣超', '自營商買賣超']]  # 三大法人買賣超

        # 合併數據
        stock_data = stock_data.join(margin_df, how='left').join(twse_data, how='left')
        stock_data.fillna(method='ffill', inplace=True)  # 前向填充缺失值
        stock_data = stock_data[['Close', 'Volume', 'MA10', 'RSI', 
                                 'margin_buy', 'margin_sell', 'short_buy', 'short_sell',
                                 '外資買賣超', '投信買賣超', '自營商買賣超']].dropna()

        # 標準化數據
        scaler = MinMaxScaler(feature_range=(0, 1))
        scaled_data = scaler.fit_transform(stock_data.values)

        # 生成數據集
        X, Y = create_dataset(scaled_data, time_step=126, future_days=120)
        if len(X) == 0:
            return None, stock_code_tw

        positive_count = np.sum(Y)
        if positive_count > 0:
            print(f"{stock_code_tw}: Found {positive_count} positive samples (50% increase)")
        else:
            print(f"{stock_code_tw}: No positive samples found")

        return (X, Y, scaler), stock_code_tw
    except Exception as e:
        print(f"Error processing {stock_code_tw}: {e}")
        return None, stock_code_tw

def compute_rsi(data, periods=14):
    delta = data.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=periods).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=periods).mean()
    rs = gain / loss
    return 100 - (100 / (1 + rs))

# 並行下載數據
all_X_train, all_Y_train, all_X_test, all_Y_test = [], [], [], []
scalers = {}
with ThreadPoolExecutor(max_workers=10) as executor:
    results = executor.map(download_stock_data, stock_codes[:200])

for result, stock_code in results:
    if result is None:
        print(f"Failed to process {stock_code}")
        continue
    X, Y, scaler = result
    scalers[stock_code] = scaler

    train_size = int(len(X) * 0.8)
    X_train, X_test = X[:train_size], X[train_size:]
    Y_train, Y_test = Y[:train_size], Y[train_size:]

    all_X_train.append(X_train)
    all_Y_train.append(Y_train)
    all_X_test.append(X_test)
    all_Y_test.append(Y_test)

# 合併數據
if all_X_train:
    X_train = np.concatenate(all_X_train, axis=0)
    Y_train = np.concatenate(all_Y_train, axis=0)
    X_test = np.concatenate(all_X_test, axis=0)
    Y_test = np.concatenate(all_Y_test, axis=0)

    print(f"Training set: {len(Y_train)} samples, Positive: {np.sum(Y_train)} ({100 * np.sum(Y_train) / len(Y_train):.2f}%)")
    print(f"Test set: {len(Y_test)} samples, Positive: {np.sum(Y_test)} ({100 * np.sum(Y_test) / len(Y_test):.2f}%)")

    # 應用 SMOTE
    smote = SMOTE(random_state=42)
    X_train_reshaped = X_train.reshape(X_train.shape[0], -1)
    X_train_resampled, Y_train_resampled = smote.fit_resample(X_train_reshaped, Y_train)
    X_train_resampled = X_train_resampled.reshape(-1, X_train.shape[1], X_train.shape[2])
    X_train = X_train_resampled
    Y_train = Y_train_resampled

    print(f"After SMOTE, Training set: {len(Y_train)} samples, Positive: {np.sum(Y_train)} ({100 * np.sum(Y_train) / len(Y_train):.2f}%)")

    X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], X_train.shape[2])
    X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], X_test.shape[2])

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    X_train = torch.from_numpy(X_train).float().to(device)
    Y_train = torch.from_numpy(Y_train).float().to(device)
    X_test = torch.from_numpy(X_test).float().to(device)
    Y_test = torch.from_numpy(Y_test).float().to(device)

    print(f"Training set: {len(Y_train)} samples, Positive: {torch.sum(Y_train).item()} ({100 * torch.sum(Y_train).item() / len(Y_train):.2f}%)")
    print(f"Test set: {len(Y_test)} samples, Positive: {torch.sum(Y_test).item()} ({100 * torch.sum(Y_test).item() / len(Y_test):.2f}%)")

    # 定義 Focal Loss
    class FocalLoss(nn.Module):
        def __init__(self, alpha=0.5, gamma=1.5):
            super(FocalLoss, self).__init__()
            self.alpha = alpha
            self.gamma = gamma

        def forward(self, inputs, targets):
            BCE_loss = nn.BCEWithLogitsLoss(reduction='none')(inputs, targets)
            pt = torch.exp(-BCE_loss)
            F_loss = self.alpha * (1 - pt) ** self.gamma * BCE_loss
            return F_loss.mean()

    # 定義 LSTM 模型
    class LSTMModel(nn.Module):
        def __init__(self, input_size=11, hidden_size=128, num_layers=3):  # 修改 input_size
            super(LSTMModel, self).__init__()
            self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True, dropout=0.4)
            self.fc = nn.Linear(hidden_size, 1)
    
        def forward(self, x):
            h_0 = torch.zeros(3, x.size(0), 128).to(x.device)
            c_0 = torch.zeros(3, x.size(0), 128).to(x.device)
            out, _ = self.lstm(x, (h_0, c_0))
            out = self.fc(out[:, -1, :])
            return out

    # 釋放內存
    torch.cuda.empty_cache()
    gc.collect()

    # 初始化模型
    model = LSTMModel().to(device)
    print(f"Y_train positive count: {torch.sum(Y_train).item()}")
    criterion = FocalLoss(alpha=0.5, gamma=1.5)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.0005, weight_decay=5e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, verbose=False)
    scaler = GradScaler()  # 混合精度訓練

    # 批量訓練
    train_dataset = TensorDataset(X_train, Y_train)
    batch_size = 32
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_dataset = TensorDataset(X_test, Y_test)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)  # 測試集分批

    best_loss = float('inf')
    patience = 20
    counter = 0
    num_epochs = 200
    accumulation_steps = 4  # 梯度累積步數

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        optimizer.zero_grad()  # 在循環開始時清零梯度
        for i, (batch_X, batch_Y) in enumerate(train_loader):
            batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
            with autocast():  # 混合精度
                outputs = model(batch_X).squeeze()
                loss = criterion(outputs, batch_Y) / accumulation_steps
            scaler.scale(loss).backward()
            if (i + 1) % accumulation_steps == 0:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()
            total_loss += loss.item() * accumulation_steps
        
        model.eval()
        val_loss = 0
        val_outputs = []
        with torch.no_grad():
            for batch_X, batch_Y in test_loader:
                batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
                with autocast():
                    outputs = model(batch_X).squeeze()
                    loss = criterion(outputs, batch_Y)
                val_outputs.append(outputs)
                val_loss += loss.item()
        
        val_outputs = torch.cat(val_outputs)
        val_loss = val_loss / len(test_loader)
        scheduler.step(val_loss)
        
        if (epoch + 1) % 10 == 0:
            avg_loss = total_loss / len(train_loader)
            print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {avg_loss:.4f}, Val Loss: {val_loss:.4f}')
        
        if val_loss < best_loss:
            best_loss = val_loss
            counter = 0
            torch.save(model.state_dict(), 'best_model.pth')
        else:
            counter += 1
            if counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # 預測與評估
    model.load_state_dict(torch.load('best_model.pth', weights_only=True))
    model.eval()
    test_predict = []
    with torch.no_grad():
        for batch_X, _ in test_loader:
            batch_X = batch_X.to(device)
            with autocast():
                outputs = model(batch_X).squeeze()
            test_predict.append(outputs.cpu())
    
    test_predict = torch.cat(test_predict)
    # 計算最佳閾值
    precision, recall, thresholds = precision_recall_curve(Y_test.cpu().numpy(), test_predict.numpy())
    f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)
    optimal_idx = np.argmax(f1_scores)
    optimal_threshold = thresholds[optimal_idx]
    print(f"Optimal threshold: {optimal_threshold:.4f}")

    test_predict_binary = (test_predict >= optimal_threshold).float()
    print(classification_report(Y_test.cpu().numpy(), test_predict_binary.numpy(), zero_division=0))
    print(f'ROC AUC: {roc_auc_score(Y_test.cpu().numpy(), test_predict.numpy()):.4f}')

    # 混淆矩陣
    cm = confusion_matrix(Y_test.cpu().numpy(), test_predict_binary.numpy())
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix')
    plt.show()

    # Precision-Recall 曲線
    plt.figure(figsize=(8, 6))
    plt.plot(recall, precision, label='Precision-Recall Curve')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curve')
    plt.legend()
    plt.show()

    # 預測概率分佈
    plt.figure(figsize=(10, 6))
    plt.hist(test_predict.numpy(), bins=50, alpha=0.5, label='Predicted Probabilities')
    plt.axvline(x=optimal_threshold, color='r', linestyle='--', label=f'Optimal Threshold ({optimal_threshold:.4f})')
    plt.xlabel('Predicted Probability')
    plt.ylabel('Frequency')
    plt.title('Distribution of Predicted Probabilities')
    plt.legend()
    plt.show()

    # 保存模型和數據
    torch.save(model.state_dict(), 'lstm_stock_model.pth')
    with open('stock_data.pkl', 'wb') as f:
        pickle.dump((X_train, Y_train, X_test, Y_test, scalers), f)

    # 清理內存
    del X_train, Y_train, X_test, Y_test, val_outputs, test_predict
    torch.cuda.empty_cache()
    gc.collect()
else:
    print("No valid data found. Consider expanding the time range or stock list.")
```
