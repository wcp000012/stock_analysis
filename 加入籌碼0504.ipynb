```python
# 這區程式碼是用來下載籌碼資料到sqllite方便之後跟技術資料合併
import requests
import io
import re
import time
import sqlite3
import pandas as pd
from datetime import datetime, timedelta
from urllib.error import HTTPError, URLError
#這段程式碼是會成功保留適當的股票代號(用JASON)，並依此merge融資融卷和外資買賣超數據
def fetch_t86_json(date_str: str) -> pd.DataFrame:
    # 建議使用 RWD JSON 介面（UTF-8）
    url = f"https://www.twse.com.tw/rwd/zh/fund/T86?response=json&date={date_str}&selectType=ALLBUT0999"
    print(url)
    j = requests.get(url, timeout=30).json()

    # 介面會回傳 fields + data
    cols = j.get("fields", [])
    data = j.get("data", [])
    if not cols or not data:
        return pd.DataFrame()

    df = pd.DataFrame(data, columns=cols)

    # 清理欄位
    df["證券代號"] = df["證券代號"].astype(str).str.strip()
    df = df[~df["證券代號"].isin(["合計", "總計"])]

    # 需要的欄位盡量彈性處理（不同日期，欄名可能有差）
    # 1) 外資兩欄分項 → 相加
    col_fx_net = None
    fx_net_no_prop = "外陸資買賣超股數(不含外資自營商)"
    fx_prop_net   = "外資自營商買賣超股數"
    fx_buy_no_prop = "外陸資買進股數(不含外資自營商)"
    fx_sell_no_prop= "外陸資賣出股數(不含外資自營商)"
    fx_prop_buy    = "外資自營商買進股數"
    fx_prop_sell   = "外資自營商賣出股數"
    #print(df.columns)
    # 轉數字（把千分位逗號拿掉）
    def to_num(s):
        return pd.to_numeric(pd.Series(s).astype(str).str.replace(',', '', regex=False), errors='coerce')

    num_cands = [c for c in df.columns if any(k in c for k in ["買進股數", "賣出股數", "買賣超股數"])]
    for c in num_cands:
        df[c] = to_num(df[c])

    if fx_net_no_prop in df.columns and fx_prop_net in df.columns:
        df["外資買賣超股數"] = df[fx_net_no_prop] + df[fx_prop_net]
        col_fx_net = "外資買賣超股數"
    elif "外資買賣超股數" in df.columns:
        # 有些日期可能直接提供彙總欄
        col_fx_net = "外資買賣超股數"
    elif all(c in df.columns for c in [fx_buy_no_prop, fx_sell_no_prop, fx_prop_buy, fx_prop_sell]):
        # 從買進/賣出去推回淨額
        df["外資買賣超股數"] = (df[fx_buy_no_prop] - df[fx_sell_no_prop] +
                            df[fx_prop_buy] - df[fx_prop_sell])
        col_fx_net = "外資買賣超股數"
    else:
        # 最少也保留個別欄位供你後處理
        pass

    # 最終只留你要的幾欄
    keep = ["證券代號", col_fx_net, "投信買賣超股數", "自營商買賣超股數"]
    keep = [c for c in keep if c and c in df.columns]
    df = df[keep].copy().set_index("證券代號")
    
    return df


def download_chip_data(date):
    date_str = date.replace('-', '')

    # ---------- 【融資融券 MI_MARGN】 ----------
    try:
        # 修正 URL：改用 & 而不是 &amp;
        twse_margin_url = (
            "https://www.twse.com.tw/exchangeReport/MI_MARGN"
            f"?response=json&date={date_str}&selectType=ALL"
        )
        #print(twse_margin_url)
        # 建議：讀取後先找出真正表頭那一段（略）
        j = requests.get(twse_margin_url, timeout=30).json()
        
        # 找到第二個表格（融資融券彙總）
        tables = j.get("tables", [])
        #print(tables)
        target_table = tables[1]  # 第二個表格
        
        cols = target_table.get("fields", [])
        data = target_table.get("data", [])
        print(cols)
        
        
        idx_margin_buy = cols.index("買進")
        idx_margin_sell = cols.index("賣出")
        idx_short_buy = cols.index("買進", idx_margin_buy + 1)
        idx_short_sell = cols.index("賣出", idx_margin_sell + 1)

        
        need_idx = [cols.index("代號"), idx_margin_buy, idx_margin_sell, idx_short_buy, idx_short_sell]
        margin_data = pd.DataFrame([[row[i] for i in need_idx] for row in data],
                          columns=["證券代號", "margin_buy", "margin_sell", "short_buy", "short_sell"])

        #margin_cols = ['代號', '買進', '賣出', '買進.1', '賣出.1']
        #margin_data = margin_data[margin_cols].copy()
        #margin_data.columns = ['證券代號', 'margin_buy', 'margin_sell', 'short_buy', 'short_sell']

        margin_data['證券代號'] = margin_data['證券代號'].astype(str).str.strip()
        margin_data = margin_data[
            margin_data['證券代號'].notna() &
            ~margin_data['證券代號'].isin(['合計', '總計'])
        ]

        for c in ['margin_buy', 'margin_sell', 'short_buy', 'short_sell']:
            margin_data[c] = pd.to_numeric(
                margin_data[c].astype(str).str.replace(',', '', regex=False),
                errors='coerce'
            )
        margin_data.set_index('證券代號', inplace=True)
        #print(margin_data.head())
    
    except Exception as e:
        print(f"{date}: Failed to fetch margin data: {e}")
        margin_data = pd.DataFrame()

    # ---------- 【三大法人 T86】（改走 JSON） ----------
    try:
        twse_data = fetch_t86_json(date_str)  # ← 用上面那段 JSON 版本
        
    except Exception as e:
        print(f"{date}: Failed to fetch TWSE T86 data: {e}")
        twse_data = pd.DataFrame()

    # ---------- 合併 ----------
    if not margin_data.empty:
        out = margin_data.join(twse_data, how='left')
        out = out.apply(lambda x: pd.to_numeric(x, errors='coerce'))
        print(f"Download data of {date} successfully. shape={out.shape}")
        return out
    else:
        print(f"Fail to download data of {date}")
        return None
    

def upsert_chip(con, df: pd.DataFrame, trade_date, ticker_col='證券代號'):
    
    """
    將籌碼 DataFrame 寫入 SQLite 的 chip_daily。
    - 支援 index 是「證券代號」的情況
    - 自動把中文欄位名稱對應到資料庫英文字段
    - 自動補齊缺欄位、轉數值型別
    - trade_date: 可傳入 datetime/date 或 'YYYY-MM-DD' 字串
    """

    if df is None or df.empty:
        return

    # 若 index 名稱是「證券代號」，先轉回欄位
    if df.index.name == ticker_col and ticker_col not in df.columns:
        df = df.reset_index()

    # 欄位對應：把資料來源的欄位名稱映射到 DB 欄位
    rename_map = {
        ticker_col: 'Ticker',
        'margin_buy': 'margin_buy',
        'margin_sell': 'margin_sell',
        'short_buy': 'short_buy',
        'short_sell': 'short_sell',
        '外資買賣超股數': 'foreign_net',
        '投信買賣超股數': 'invest_net',
        '自營商買賣超股數': 'dealer_net',
    }
    df = df.rename(columns=rename_map)

    # 加入 Date 欄位（全表同一天）
    df['Date'] = pd.to_datetime(trade_date).date()
    
    # 需要的欄位
    required_cols = [
        'Date', 'Ticker',
        'margin_buy', 'margin_sell',
        'short_buy', 'short_sell',
        'foreign_net', 'invest_net', 'dealer_net'
    ]
    
    # 只保留需要的欄位順序，同時可以確保dataframe的順序有和sqllite insert的順序一樣
    df = df[required_cols]


    rows = df.values.tolist()
    cur = con.cursor()
    cur.executemany(
        """
        INSERT INTO chip_daily (Date, Ticker, margin_buy, margin_sell, short_buy, short_sell, foreign_net, invest_net, dealer_net)
        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
        ON CONFLICT(Date, Ticker) DO UPDATE SET
            margin_buy=excluded.margin_buy,
            margin_sell=excluded.margin_sell,
            short_buy=excluded.short_buy,
            short_sell=excluded.short_sell,
            foreign_net=excluded.foreign_net,
            invest_net=excluded.invest_net,
            dealer_net=excluded.dealer_net;
        """
    , rows)
    con.commit()

def daterange(start_date: str, end_date: str):
    d0 = datetime.strptime(start_date, '%Y-%m-%d').date()
    d1 = datetime.strptime(end_date, '%Y-%m-%d').date()
    d = d0
    while d <= d1:
        yield d.isoformat()
        d += timedelta(days=1)

def init_sqlite(db_path='chip.db'):
    con = sqlite3.connect(db_path)
    cur = con.cursor()
    cur.execute(
        """
        CREATE TABLE IF NOT EXISTS chip_daily (
            Date DATE NOT NULL,
            Ticker TEXT NOT NULL,
            margin_buy REAL,
            margin_sell REAL,
            short_buy REAL,
            short_sell REAL,
            foreign_net REAL,
            invest_net REAL,
            dealer_net REAL,
            PRIMARY KEY (Date, Ticker)
        );
        """
    )
    cur.execute('CREATE INDEX IF NOT EXISTS idx_chip_ticker_date ON chip_daily (Ticker, Date);')
    con.commit()
    return con

def download_chip_range_to_sqlite(start_date: str, end_date: str, db_path='chip.db', sleep_sec=0.8, max_retry=3):
    con = init_sqlite(db_path)
    try:
        for ds in daterange(start_date, end_date):
            # 略過已入庫的日期（簡化：查是否存在任一筆）
            existed = pd.read_sql_query('SELECT 1 FROM chip_daily WHERE Date=? LIMIT 1', con, params=(ds,))
            if not existed.empty:
                continue

            for attempt in range(1, max_retry+1):
                try:
                    df = download_chip_data(ds)
                    if not df.empty:
                        upsert_chip(con, df, ds)
                        print(f'[OK] {ds} rows={len(df)}')
                    else:
                        print(f'[WARN] {ds} 無資料或非交易日')
                    break
                except (HTTPError, URLError, TimeoutError) as nete:
                    print(f'[RETRY {attempt}] {ds} 網路錯誤：{nete}')
                    time.sleep(sleep_sec * attempt)
                except Exception as e:
                    print(f'[FAIL] {ds} 解析失敗：{e}')
                    break
            time.sleep(sleep_sec)
    finally:
        con.close()
#download_chip_range_to_sqlite('2015-01-01','2024-12-31')
download_chip_range_to_sqlite('2017-05-01','2025-01-01')

#download_chip_data('2017-04-10')
#download_chip_data('2024-12-31')


```

```python
# 2025.12.14嘗試大量merge資料開始訓練但遇到sqlite3.connection不iterable的問題造成無法整個資料整理程序使用多執行續提升效率，跟AI討論後可能採取作法是多執行續下載，但是merge食用單執行續
# 2025.12.20 後來詢問CHatGPT5，按照建議開啟多個db connection，並且修掉minmaxscaler處理date可能遇到的問題，但出現另一個error
# 2026.01.04 之前的error原因是因為executer.map會自動去iterate給進去的東西給不同的worker去執行函數，丟純字串會變成一個字母一個字母去iterate，現在解掉這個問題後發現他會下載並merge data但是好像沒有在進行training。還有另一個小問題是使用multiple worker時，疑似會有額外的error"Error processing 1219.TW: Cannot set a DataFrame with multiple columns to the single column MA10"
# 2026.01.13 解決部會traning的問題了，應該是UnBoundLocalError=>"在 try 裡某些早退（例如 return None, stock_code_tw）或異常發生之前，sql_con 尚未賦值；"，現在在training階段遇到新的error，最下面有copilot的建議
# 2026.01.18 解決training階段遇到的error，因為加入基數資料所以input size變了，model已改成隨著input_size動態init，但現在感覺跑起來很慢，而且copilot似乎有一些對SMOTE方法的建議。下一次可以先看一下怎麼使用GPU來幫忙training。
import sqlite3
import yfinance as yf
import pandas as pd
import numpy as np
import requests
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, precision_recall_curve
import matplotlib.pyplot as plt
import seaborn as sns
from concurrent.futures import ThreadPoolExecutor
import pickle
from torch.utils.data import TensorDataset, DataLoader
import gc
from imblearn.over_sampling import SMOTE
import os
from torch.cuda.amp import autocast, GradScaler


# 設置環境變數以優化內存分配
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# 獲取台灣股票列表
url = 'https://isin.twse.com.tw/isin/C_public.jsp?strMode=2'
response = requests.get(url)
response.encoding = 'big5'
dfs = pd.read_html(response.text)
stock_list = dfs[0]
stock_list.columns = stock_list.iloc[0]
stock_list = stock_list[1:]
stock_list = stock_list[stock_list['有價證券代號及名稱'].str.contains(r'\d{4}\s+')]
stock_list['股票代號'] = stock_list['有價證券代號及名稱'].str.split().str[0]
stock_list['股票名稱'] = stock_list['有價證券代號及名稱'].str.split().str[1]
stock_list = stock_list[['股票代號', '股票名稱']]
stock_codes = stock_list['股票代號'].tolist()

def query_chip(con, ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
    q = """
    SELECT Date, Ticker, margin_buy, margin_sell, short_buy, short_sell,
           foreign_net, invest_net, dealer_net
    FROM chip_daily
    WHERE Ticker = ? AND Date BETWEEN ? AND ?
    ORDER BY Date
    """
    df = pd.read_sql_query(q, con, params=(ticker, start_date, end_date), parse_dates=['Date'])
    return df

# 定義數據集生成函數（分類任務）
def create_dataset(data,close_raw, time_step=126, future_days=120):
    X, Y = [], []
    for i in range(len(data) - time_step - future_days):
        X.append(data[i:(i + time_step), :])
        current_price = close_raw[i + time_step]
        future_price = close_raw[i + time_step + future_days]
        price_change = (future_price - current_price) / current_price
        Y.append(1 if price_change >= 0.5 else 0)  # 維持 50% 閾值
    return np.array(X), np.array(Y)

# 定義下載單支股票數據的函數



def download_stock_data(stock_code, DB_PATH = "chip.db"):
    try:
        #print("start to download stock "+str(stock_code))
        stock_code_tw = stock_code + ".TW"
        start_date = '2015-01-01'
        end_date = '2024-12-31'
        
        # 下載價格數據 (yfinance)
        stock_data = yf.download(stock_code_tw, start=start_date, end=end_date, progress=False)
        if stock_data.empty or len(stock_data) < 120 + 126:
            return None, stock_code_tw

        # 計算技術指標
        stock_data['Volume'] = stock_data['Volume']
        stock_data['MA10'] = stock_data['Close'].rolling(window=10).mean()
        stock_data['RSI'] = compute_rsi(stock_data['Close'], 14)
        #print("DB_PATH="+str(DB_PATH))
        
        with sqlite3.connect(DB_PATH) as sql_con:
            chip = query_chip(sql_con, stock_code, start_date, end_date)

        #sql_con = sqlite3.connect(DB_PATH)
        #chip=query_chip(sql_con,stock_code,start_date,end_date)
        #print("Process "+str(stock_code)+" successfully!")

        # 第一個 DataFrame：stock_data
        stock_data.index = pd.to_datetime(stock_data.index)
        stock_data.columns = stock_data.columns.get_level_values(0)  # 去掉 MultiIndex 第二層
        stock_data = stock_data.reset_index()  # 把 Date 從 index 變成欄位

        # 第二個 DataFrame：chip
        chip['Date'] = pd.to_datetime(chip['Date'])
        

        # 合併（只用 Date）
        stock_data_merged = pd.merge(
            stock_data,
            chip.drop(columns=['Ticker']),  # 如果不需要 Ticker，可以直接丟掉
            on='Date',
            how='inner'  # 或 'left' 視需求
        ).drop(columns=['Date'])
        stock_data_merged=stock_data_merged.dropna()
        close_raw = stock_data_merged["Close"].to_numpy().astype(float)
        
        scaler = MinMaxScaler(feature_range=(0, 1))
        feature_cols = [
            "Close","High","Low","Open","Volume","MA10","RSI",
            "margin_buy","margin_sell","short_buy","short_sell",
            "foreign_net","invest_net","dealer_net"
        ]
        #print(stock_data_merged.head())
        scaled_data = scaler.fit_transform(stock_data_merged[feature_cols].to_numpy())

        # 生成數據集
        X, Y = create_dataset(scaled_data, close_raw, time_step=126, future_days=120)
        if len(X) == 0:
            return None, stock_code_tw

        positive_count = np.sum(Y)
        if positive_count > 0:
            print(f"{stock_code_tw}: Found {positive_count} positive samples (50% increase)")
        else:
            print(f"{stock_code_tw}: No positive samples found")

        return (X, Y, scaler), stock_code_tw
    except Exception as e:
        print(f"Error processing {stock_code_tw}: {e}")
        return None, stock_code_tw
    
    #finally:
    #    sql_con.close()


def compute_rsi(data, periods=14):
    delta = data.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=periods).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=periods).mean()
    rs = gain / loss
    return 100 - (100 / (1 + rs))



# 並行下載數據
all_X_train, all_Y_train, all_X_test, all_Y_test = [], [], [], []
scalers = {}
with ThreadPoolExecutor(max_workers=10) as executor:
    results = executor.map(download_stock_data, stock_codes[:20])

for result, stock_code in results:
    if result is None:
        print(f"Failed to process {stock_code}")
        continue
    X, Y, scaler = result
    scalers[stock_code] = scaler

    train_size = int(len(X) * 0.8)
    X_train, X_test = X[:train_size], X[train_size:]
    Y_train, Y_test = Y[:train_size], Y[train_size:]

    all_X_train.append(X_train)
    all_Y_train.append(Y_train)
    all_X_test.append(X_test)
    all_Y_test.append(Y_test)

#print("all_X_train=")
#print(all_X_train)
num_features = X_train.shape[2]
    
# 合併數據
if all_X_train:
    X_train = np.concatenate(all_X_train, axis=0)
    Y_train = np.concatenate(all_Y_train, axis=0)
    X_test = np.concatenate(all_X_test, axis=0)
    Y_test = np.concatenate(all_Y_test, axis=0)

    print(f"Training set: {len(Y_train)} samples, Positive: {np.sum(Y_train)} ({100 * np.sum(Y_train) / len(Y_train):.2f}%)")
    print(f"Test set: {len(Y_test)} samples, Positive: {np.sum(Y_test)} ({100 * np.sum(Y_test) / len(Y_test):.2f}%)")

    # 應用 SMOTE
    smote = SMOTE(random_state=42)
    X_train_reshaped = X_train.reshape(X_train.shape[0], -1)
    X_train_resampled, Y_train_resampled = smote.fit_resample(X_train_reshaped, Y_train)
    X_train_resampled = X_train_resampled.reshape(-1, X_train.shape[1], X_train.shape[2])
    X_train = X_train_resampled
    Y_train = Y_train_resampled

    print(f"After SMOTE, Training set: {len(Y_train)} samples, Positive: {np.sum(Y_train)} ({100 * np.sum(Y_train) / len(Y_train):.2f}%)")

    X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], X_train.shape[2])
    X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], X_test.shape[2])

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    X_train = torch.from_numpy(X_train).float().to(device)
    Y_train = torch.from_numpy(Y_train).float().to(device)
    X_test = torch.from_numpy(X_test).float().to(device)
    Y_test = torch.from_numpy(Y_test).float().to(device)

    print(f"Training set: {len(Y_train)} samples, Positive: {torch.sum(Y_train).item()} ({100 * torch.sum(Y_train).item() / len(Y_train):.2f}%)")
    print(f"Test set: {len(Y_test)} samples, Positive: {torch.sum(Y_test).item()} ({100 * torch.sum(Y_test).item() / len(Y_test):.2f}%)")

    # 定義 Focal Loss
    class FocalLoss(nn.Module):
        def __init__(self, alpha=0.5, gamma=1.5):
            super(FocalLoss, self).__init__()
            self.alpha = alpha
            self.gamma = gamma

        def forward(self, inputs, targets):
            BCE_loss = nn.BCEWithLogitsLoss(reduction='none')(inputs, targets)
            pt = torch.exp(-BCE_loss)
            F_loss = self.alpha * (1 - pt) ** self.gamma * BCE_loss
            return F_loss.mean()

    # 定義 LSTM 模型
    class LSTMModel(nn.Module):
        def __init__(self, input_size: int, hidden_size: int=128, num_layers:int =3, dropout: float=0.4):  # 修改 input_size
            super(LSTMModel, self).__init__()
            self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size, num_layers=num_layers, batch_first=True, dropout=dropout if num_layers > 1 else 0.0)
            self.fc = nn.Linear(hidden_size, 1)
    
        def forward(self, x):
            h_0 = torch.zeros(3, x.size(0), 128).to(x.device)
            c_0 = torch.zeros(3, x.size(0), 128).to(x.device)
            out, _ = self.lstm(x, (h_0, c_0))
            out = self.fc(out[:, -1, :])
            return out

    # 釋放內存
    torch.cuda.empty_cache()
    gc.collect()

    # 初始化模型
    model = LSTMModel(input_size=num_features, hidden_size=128, num_layers=3, dropout=0.4).to(device)
    print(f"Y_train positive count: {torch.sum(Y_train).item()}")
    criterion = FocalLoss(alpha=0.5, gamma=1.5)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.0005, weight_decay=5e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, verbose=False)
    scaler = GradScaler()  # 混合精度訓練

    # 批量訓練
    train_dataset = TensorDataset(X_train, Y_train)
    batch_size = 32
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_dataset = TensorDataset(X_test, Y_test)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)  # 測試集分批

    best_loss = float('inf')
    patience = 20
    counter = 0
    num_epochs = 200
    accumulation_steps = 4  # 梯度累積步數

    for epoch in range(num_epochs):
        model.train()
        total_loss = 0
        optimizer.zero_grad()  # 在循環開始時清零梯度
        for i, (batch_X, batch_Y) in enumerate(train_loader):
            batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
            with autocast():  # 混合精度
                outputs = model(batch_X).squeeze()
                loss = criterion(outputs, batch_Y) / accumulation_steps
            scaler.scale(loss).backward()
            if (i + 1) % accumulation_steps == 0:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()
            total_loss += loss.item() * accumulation_steps
        
        model.eval()
        val_loss = 0
        val_outputs = []
        with torch.no_grad():
            for batch_X, batch_Y in test_loader:
                batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
                with autocast():
                    outputs = model(batch_X).squeeze()
                    loss = criterion(outputs, batch_Y)
                val_outputs.append(outputs)
                val_loss += loss.item()
        
        val_outputs = torch.cat(val_outputs)
        val_loss = val_loss / len(test_loader)
        scheduler.step(val_loss)
        
        if (epoch + 1) % 10 == 0:
            avg_loss = total_loss / len(train_loader)
            print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {avg_loss:.4f}, Val Loss: {val_loss:.4f}')
        
        if val_loss < best_loss:
            best_loss = val_loss
            counter = 0
            torch.save(model.state_dict(), 'best_model.pth')
        else:
            counter += 1
            if counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # 預測與評估
    model.load_state_dict(torch.load('best_model.pth', weights_only=True))
    model.eval()
    test_predict = []
    with torch.no_grad():
        for batch_X, _ in test_loader:
            batch_X = batch_X.to(device)
            with autocast():
                outputs = model(batch_X).squeeze()
            test_predict.append(outputs.cpu())
    
    test_predict = torch.cat(test_predict)
    # 計算最佳閾值
    precision, recall, thresholds = precision_recall_curve(Y_test.cpu().numpy(), test_predict.numpy())
    f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)
    optimal_idx = np.argmax(f1_scores)
    optimal_threshold = thresholds[optimal_idx]
    print(f"Optimal threshold: {optimal_threshold:.4f}")

    test_predict_binary = (test_predict >= optimal_threshold).float()
    print(classification_report(Y_test.cpu().numpy(), test_predict_binary.numpy(), zero_division=0))
    print(f'ROC AUC: {roc_auc_score(Y_test.cpu().numpy(), test_predict.numpy()):.4f}')

    # 混淆矩陣
    cm = confusion_matrix(Y_test.cpu().numpy(), test_predict_binary.numpy())
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title('Confusion Matrix')
    plt.show()

    # Precision-Recall 曲線
    plt.figure(figsize=(8, 6))
    plt.plot(recall, precision, label='Precision-Recall Curve')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curve')
    plt.legend()
    plt.show()

    # 預測概率分佈
    plt.figure(figsize=(10, 6))
    plt.hist(test_predict.numpy(), bins=50, alpha=0.5, label='Predicted Probabilities')
    plt.axvline(x=optimal_threshold, color='r', linestyle='--', label=f'Optimal Threshold ({optimal_threshold:.4f})')
    plt.xlabel('Predicted Probability')
    plt.ylabel('Frequency')
    plt.title('Distribution of Predicted Probabilities')
    plt.legend()
    plt.show()

    # 保存模型和數據
    torch.save(model.state_dict(), 'lstm_stock_model.pth')
    with open('stock_data.pkl', 'wb') as f:
        pickle.dump((X_train, Y_train, X_test, Y_test, scalers), f)

    # 清理內存
    del X_train, Y_train, X_test, Y_test, val_outputs, test_predict
    torch.cuda.empty_cache()
    gc.collect()
else:
    print("No valid data found. Consider expanding the time range or stock list.")
```

```python
#2026/1/20 為了使用machine上的GPU，必須回去使用default的python environment，所以先把資料下載下來，但這次會先存成parquet格式，因為之後可以直接讀取成dataframe。現在遇到的困難是我需要在自己的kernel和machine上的kernel安裝一個librery，下次再弄
#2026/1/25 安裝pyarrow解決第一個問題，但遇到另一個"而 PyArrow 會在第一次寫檔時動態註冊某些擴充型別（例如 pandas.period）。這個註冊表是全域而且非 thread-safe，因此不同 thread 幾乎同時註冊或查找時，就可能一個 thread 尚未註冊完成、另一個卻偵測到「已註冊」、或找不到而報錯，於是你就看到了這兩種互斥的錯誤訊息"，拿程式碼焊error問GPT即可
#2026/2/16 用自己寫的write_parquet_via_pyarrow可以避免多重註冊問題，但之後讀取還是要跟copilot確認一下。針對yfinance遇到快速的request會回傳多重stock種類的table的問題也有檢查機制了，目前已經大部分可成功下載，但還是有小部分會失敗，可能有些太新的股票並沒有辦法追朔到2015，之後可以考慮先用目前的data讀取然後做training
import time
import random
import os
import json
import yfinance as yf
import pandas as pd
import sqlite3
from concurrent.futures import ThreadPoolExecutor
import requests
from datetime import datetime

url = 'https://isin.twse.com.tw/isin/C_public.jsp?strMode=2'
response = requests.get(url)
response.encoding = 'big5'
dfs = pd.read_html(response.text)
stock_list = dfs[0]
stock_list.columns = stock_list.iloc[0]
stock_list = stock_list[1:]
stock_list = stock_list[stock_list['有價證券代號及名稱'].str.contains(r'\d{4}\s+')]
stock_list['股票代號'] = stock_list['有價證券代號及名稱'].str.split().str[0]
stock_list['股票名稱'] = stock_list['有價證券代號及名稱'].str.split().str[1]
stock_list = stock_list[['股票代號', '股票名稱']]
stock_codes = stock_list['股票代號'].tolist()

##################################################################


def debug_dump_df(df: pd.DataFrame, tag: str, out_dir: str = "debug_snapshots"):
    """把當下 df 的結構與前幾列存檔，不用 print（避免時序變化）。"""
    os.makedirs(out_dir, exist_ok=True)
    ts = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
    base = os.path.join(out_dir, f"{tag}_{ts}")

    # 1) 欄位資訊（支援 MultiIndex）
    if isinstance(df.columns, pd.MultiIndex):
        cols = [tuple(map(str, lvl)) for lvl in df.columns.tolist()]
        cols_info = {"type": "MultiIndex", "columns": cols}
    else:
        cols_info = {"type": "Index", "columns": list(map(str, df.columns.tolist()))}

    # 2) 索引資訊
    idx_info = {
        "is_multiindex": isinstance(df.index, pd.MultiIndex),
        "index_name": str(df.index.name),
        "len": len(df.index)
    }

    # 3) dtypes
    dtypes_info = {str(c): str(t) for c, t in df.dtypes.items()}

    meta = {
        "cols_info": cols_info,
        "index_info": idx_info,
        "dtypes": dtypes_info,
        "shape": list(df.shape)
    }

    with open(base + "_meta.json", "w", encoding="utf-8") as f:
        json.dump(meta, f, ensure_ascii=False, indent=2)

    # 4) 前幾列（不 print，存 CSV）
    df.head(10).to_csv(base + "_head.csv", index=True, encoding="utf-8-sig")

    # 5) 也可整份存一份 parquet/feather 以便離線重現
    try:
        import pyarrow as pa, pyarrow.parquet as pq
        table = pa.Table.from_pandas(df, preserve_index=True)
        pq.write_table(table, base + "_full.parquet", compression="zstd")
    except Exception:
        # 若環境一時缺套件，至少保住 CSV
        df.to_csv(base + "_full_fallback.csv", index=True, encoding="utf-8-sig")



def _get_close_series_strict(df: pd.DataFrame, tag: str) -> pd.Series:
    """
    穩健地擷取單一 Close 欄位；若取得結果不是 Series（代表多欄或不一致），
    先做一次 debug_dump 再拋例外。
    """
    # MultiIndex 欄位：取 level=0 == 'Close'
    if isinstance(df.columns, pd.MultiIndex):
        try:
            sub = df.xs('Close', axis=1, level=0)
        except KeyError:
            debug_dump_df(df, tag + "_NoCloseLevel0")
            raise KeyError("DataFrame 沒有 'Close' (level=0) 欄位")

        if isinstance(sub, pd.DataFrame):
            if sub.shape[1] != 1:
                # 多個 'Close' 欄，存檔再報
                debug_dump_df(df, tag + "_MultiCloseMI")
                raise ValueError(f"'Close' 在 MultiIndex 下有 {sub.shape[1]} 欄，非單一 Series")
            return sub.iloc[:, 0]
        else:
            return sub

    # 單層欄位：可能有重複 'Close'
    close_cols = [c for c in df.columns if c == 'Close']
    if not close_cols:
        debug_dump_df(df, tag + "_NoClose")
        raise KeyError("DataFrame 不包含 'Close' 欄位")

    if len(close_cols) > 1:
        # 多個同名 'Close'，存檔再報
        debug_dump_df(df, tag + "_MultiCloseFlat")
        raise ValueError(f"發現 {len(close_cols)} 個 'Close' 欄，非單一 Series")

    # 一個 'Close'，但保險檢查取值型態
    sub = df['Close']
    if isinstance(sub, pd.DataFrame):
        if sub.shape[1] != 1:
            debug_dump_df(df, tag + "_CloseIsDF_MultiCols")
            raise ValueError("'Close' 取到的是多欄 DataFrame")
        return sub.iloc[:, 0]
    return sub

##################################################################

def _is_single_close_series(df: pd.DataFrame) -> bool:
    """檢查 df 的 Close 是否為單一 Series（無論原本是否是 MultiIndex）。"""
    if df is None or df.empty:
        return False

    # 情況一：單層欄位
    if not isinstance(df.columns, pd.MultiIndex):
        # 是否有且僅有一個 'Close' 欄位
        close_cols = [c for c in df.columns if c == 'Close']
        if len(close_cols) != 1:
            return False
        # 確保取到的是 Series（不是 DataFrame）
        return not isinstance(df['Close'], pd.DataFrame)

    # 情況二：雙層欄位
    lvl0 = df.columns.get_level_values(0)
    lvl1 = df.columns.get_level_values(1)

    # (Field, Ticker)
    if 'Close' in lvl0:
        sub = df.xs('Close', axis=1, level=0)
        # 只在寬度為 1（單一 ticker）時判定為單一 Series
        return isinstance(sub, pd.Series) or (isinstance(sub, pd.DataFrame) and sub.shape[1] == 1)

    # (Ticker, Field) 的話，第 1 層才是 Field
    if 'Close' in lvl1:
        sub = df.xs('Close', axis=1, level=1)
        return isinstance(sub, pd.Series) or (isinstance(sub, pd.DataFrame) and sub.shape[1] == 1)

    return False



def _normalize_to_single_ticker(df: pd.DataFrame, ticker: str) -> pd.DataFrame:
    """
    將 yfinance 回傳結果規範為單一 ticker、單層欄位。
    若本來就是單層就原樣返回；若是 MultiIndex，嘗試 (Field, Ticker) 與 (Ticker, Field) 兩種切法。
    """
    if df is None or df.empty:
        return df
    if not isinstance(df.columns, pd.MultiIndex):
        return df

    lvl0 = df.columns.get_level_values(0)
    lvl1 = df.columns.get_level_values(1)

    # (Field, Ticker)
    if ticker in lvl1:
        sub = df.xs(ticker, axis=1, level=1)
        sub.columns = sub.columns.get_level_values(0)
        return sub

    # (Ticker, Field)
    if ticker in lvl0:
        sub = df.xs(ticker, axis=1, level=0)
        sub.columns = sub.columns.get_level_values(1)
        return sub

    # 如果都不是，原樣回傳，交給上層判斷重抓
    return df


def download_single_ticker_until_ok(
    ticker: str,
    start: str,
    end: str,
    max_tries: int = 4,
    backoff_base: float = 0.5,
    auto_adjust: bool = False
) -> pd.DataFrame:
    """
    while 迴圈：只要檢測到「Close 不是單一 Series」就重抓，直到成功或超過 max_tries。
    最後一次失敗時 fallback 走 Ticker.history()。
    """
    last_err = None

    for attempt in range(1, max_tries + 1):
        try:
            df = yf.download(
                ticker,
                start=start,
                end=end,
                progress=False,
                threads=False,        # 外面如果有 ThreadPool，這裡必須 False
                group_by="column",
                auto_adjust=auto_adjust
            )

            if df is None or df.empty:
                raise RuntimeError("yfinance 無資料")

            # 規範欄位為單一 ticker、單層欄位
            df = _normalize_to_single_ticker(df, ticker)

            # 驗證 Close 是否為單一 Series
            if not _is_single_close_series(df):
                raise ValueError("Close 非單一 Series（可能是多檔、多欄或 MultiIndex 未切乾淨）")

            # 到這裡表示「可安全計算 MA / RSI」
            df.index = pd.to_datetime(df.index)
            if df.index.has_duplicates:
                df = df[~df.index.duplicated(keep='last')]
            return df

        except Exception as e:
            last_err = e
            # 最後一次嘗試 → fallback
            if attempt == max_tries:
                try:
                    ssn = requests.Session()
                    t = yf.Ticker(ticker, session=ssn)
                    df = t.history(start=start, end=end, auto_adjust=auto_adjust, actions=False)
                    if df is None or df.empty:
                        raise RuntimeError("Ticker.history() 亦無資料")
                    # 規範欄位
                    keep = [c for c in ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume'] if c in df.columns]
                    df = df[keep]
                    # 再驗證一次
                    if 'Close' not in df.columns:
                        raise ValueError("Fallback history() 無 Close 欄位")
                    df.index = pd.to_datetime(df.index)
                    if df.index.has_duplicates:
                        df = df[~df.index.duplicated(keep='last')]
                    return df
                except Exception as e2:
                    last_err = e2
                    break  # 跳出 for，交給呼叫端處理
            else:
                # 退避 + 抖動
                sleep_s = backoff_base * (2 ** (attempt - 1)) + random.uniform(0.05, 0.3)
                time.sleep(sleep_s)

    raise RuntimeError(f"下載 {ticker} 失敗：{last_err}")


def _warm_up_pyarrow():
    import pyarrow as pa  # 觸發初始化
    tmp = pd.DataFrame({
        "Date": pd.to_datetime(["2000-01-01"]),
        "x": [1.0]
    })
    tmp.to_parquet("_warmup.parquet", engine="pyarrow", index=False)
    os.remove("_warmup.parquet")

_warm_up_pyarrow()


def write_parquet_via_pyarrow(df: pd.DataFrame, out_path: str, compression: str = "zstd"):
    import pyarrow as pa
    import pyarrow.parquet as pq
    # 將 pandas DataFrame 轉成 Arrow Table，不保留 index
    table = pa.Table.from_pandas(df, preserve_index=False)
    pq.write_table(table, out_path, compression=compression)


def query_chip(con, ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
    q = """
    SELECT Date, Ticker, margin_buy, margin_sell, short_buy, short_sell,
           foreign_net, invest_net, dealer_net
    FROM chip_daily
    WHERE Ticker = ? AND Date BETWEEN ? AND ?
    ORDER BY Date
    """
    df = pd.read_sql_query(q, con, params=(ticker, start_date, end_date), parse_dates=['Date'])
    return df

def compute_rsi(data, periods=14):
    delta = data.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=periods).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=periods).mean()
    rs = gain / loss
    return 100 - (100 / (1 + rs))

def prepare_stock_to_parquet(
    stock_code,
    DB_PATH="chip.db",
    OUT_DIR="parquet_data",
    start_date="2015-01-01",
    end_date="2024-12-31"
):
    try:
        os.makedirs(OUT_DIR, exist_ok=True)

        stock_code_tw = stock_code + ".TW"

        # 1️⃣ yfinance
        
        stock_data = download_single_ticker_until_ok(stock_code_tw, start_date, end_date)
        
        if stock_data.empty or len(stock_data) < (120 + 126):
            return None

        # 2️⃣ 技術指標    
        stock_data['MA10'] = stock_data['Close'].rolling(10).mean()
        stock_data['RSI'] = compute_rsi(stock_data['Close'], 14)

        # index → Date
        stock_data.index = pd.to_datetime(stock_data.index)
        stock_data.columns = stock_data.columns.get_level_values(0)
        stock_data = stock_data.reset_index()

        # 3️⃣ chip data
        with sqlite3.connect(DB_PATH) as sql_con:
            chip = query_chip(sql_con, stock_code, start_date, end_date)

        chip['Date'] = pd.to_datetime(chip['Date'])

        # 4️⃣ merge
        df = pd.merge(
            stock_data,
            chip.drop(columns=["Ticker"]),
            on="Date",
            how="inner"
        ).dropna()
        
        
        # 5️⃣ Parquet（只存「處理後可訓練」資料）
        out_path = os.path.join(
            OUT_DIR,
            f"{stock_code_tw}_2015_2024.parquet"
        )

        write_parquet_via_pyarrow(df, out_path, compression="zstd")

        print(f"✅ Saved {stock_code_tw}: {out_path}")
        return out_path

    except Exception as e:
        print(f"❌ {stock_code_tw}: {e}")
        return None



with ThreadPoolExecutor(max_workers=10) as executor:
    list(executor.map(
        prepare_stock_to_parquet,
        #stock_codes[:20]
        stock_codes
    ))
```

```python
# 2026.03.01 現在開始讀取之前下載的資料，但是只要開始做資料處理好像就會當機，需要debug一下看是哪邊出問題
# 2026.03.07 1. 後來發現因為讀取這麼多資料很花記憶體，之前又因為一些奇怪原因default都會選到記憶體比較差的機器，之後記得選mem比較多的機器，然後用terminal看一下記憶體使用量
#             2. 發現即使用免費機器裡最好的也只有到40G ram，很容易就不夠用了。要花錢的機台最多好像也只有到90G
import glob
import torch.nn as nn
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import torch
from torch.utils.data import IterableDataset, DataLoader

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

def create_dataset(data,close_raw, time_step=126, future_days=120):
    X, Y = [], []
    for i in range(len(data) - time_step - future_days):
        X.append(data[i:(i + time_step), :])
        current_price = close_raw[i + time_step]
        future_price = close_raw[i + time_step + future_days]
        price_change = (future_price - current_price) / current_price
        Y.append(1 if price_change >= 0.5 else 0)  # 維持 50% 閾值
    return np.array(X), np.array(Y)

def load_parquet_and_build_dataset(
    parquet_path,
    time_step=126,
    future_days=120
):
    df = pd.read_parquet(parquet_path, engine="pyarrow")
    if df.empty:
        return None
    
    df.drop(columns=['Date'], inplace=True)
    close_raw = df["Close"].to_numpy().astype(float)

    scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_data = scaler.fit_transform(df[FEATURE_COLS].to_numpy())
    
    X, Y = create_dataset(
        scaled_data,
        close_raw,
        time_step=time_step,
        future_days=future_days
    )
    
    if len(X) == 0:
        return None

    return X, Y, scaler


FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]


class StockIterableDataset(IterableDataset):
    def __init__(self, parquet_files, time_step=126, future_days=120, train=True, train_split=0.8, use_log_return=True):
        self.files = parquet_files
        self.time_step = time_step
        self.future_days = future_days
        self.train = train
        self.train_split = train_split
        self.use_log_return = use_log_return

    def __iter__(self):
        for pq in self.files:
            try:
                df = pd.read_parquet(pq, engine="pyarrow")
                if df.empty:
                    continue

                # 只保留需要欄位
                drop_cols = [c for c in df.columns if c not in FEATURE_COLS and c != "Date"]
                if drop_cols: df.drop(columns=drop_cols, inplace=True)
                if "Date" in df.columns: df.drop(columns=["Date"], inplace=True)

                # downcast
                for col in FEATURE_COLS:
                    if col in df.columns:
                        if pd.api.types.is_float_dtype(df[col]):
                            df[col] = pd.to_numeric(df[col], downcast='float')
                        elif pd.api.types.is_integer_dtype(df[col]):
                            df[col] = pd.to_numeric(df[col], downcast='integer')

                close_raw = df["Close"].to_numpy(dtype=np.float32)
                feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)

                n = len(feats)
                cut = int(n * self.train_split)
                feats_part = feats[:cut] if self.train else feats[cut:]
                if len(feats_part) < self.time_step + self.future_days + 1:
                    del df, feats, feats_part, close_raw
                    gc.collect()
                    continue

                scaler = MinMaxScaler()
                scaler.fit(feats_part)
                feats_scaled = scaler.transform(feats)

                start = 0 if self.train else cut
                end = (cut if self.train else n) - self.future_days

                for i in range(start, end - self.time_step):
                    x = feats_scaled[i:i+self.time_step, :]
                    current_price = float(close_raw[i + self.time_step])
                    future_price  = float(close_raw[i + self.time_step + self.future_days])
                    if self.use_log_return:
                        y = np.log((future_price + 1e-12) / (current_price + 1e-12))
                    else:
                        y = (future_price - current_price) / (current_price + 1e-12)

                    yield torch.from_numpy(x.copy()).float(), torch.tensor(float(y), dtype=torch.float32)

                del df, feats, feats_part, feats_scaled, close_raw
                gc.collect()

            except Exception as e:
                print(f"[WARN] skip {pq}: {e}")
                continue


# 定義 LSTM 模型
class LSTMModel(nn.Module):
    def __init__(self, input_size: int, hidden_size: int=128, num_layers:int =3, dropout: float=0.4):  # 修改 input_size
        super(LSTMModel, self).__init__()
        self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size, num_layers=num_layers, batch_first=True, dropout=dropout if num_layers > 1 else 0.0)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        h_0 = torch.zeros(3, x.size(0), 128).to(x.device)
        c_0 = torch.zeros(3, x.size(0), 128).to(x.device)
        out, _ = self.lstm(x, (h_0, c_0))
        out = self.fc(out[:, -1, :])
        return out
    


all_X_train, all_Y_train = [], []
all_X_test, all_Y_test = [], []
scalers = {}

parquet_files = glob.glob("parquet_data/*.parquet")




num_features = len(FEATURE_COLS)

#print(f"Training set: {len(Y_train)} samples, Positive: {np.sum(Y_train)} ({100 * np.sum(Y_train) / len(Y_train):.2f}%)")
#print(f"Test set: {len(Y_test)} samples, Positive: {np.sum(Y_test)} ({100 * np.sum(Y_test) / len(Y_test):.2f}%)")


train_ds = StockIterableDataset(parquet_files, time_step=126, future_days=120, train=True)
test_ds  = StockIterableDataset(parquet_files, time_step=126, future_days=120, train=False)

train_loader = DataLoader(train_ds, batch_size=32, num_workers=0, pin_memory=True)  # num_workers>0 時注意 Windows/雲端相容性
test_loader  = DataLoader(test_ds,  batch_size=32, num_workers=0, pin_memory=True)

# 釋放內存
torch.cuda.empty_cache()
gc.collect()

# 初始化模型
model = LSTMModel(input_size=num_features, hidden_size=128, num_layers=3, dropout=0.4).to(device)

# 回歸損失：Huber（SmoothL1）較耐 outlier；或用 nn.MSELoss()
criterion = nn.SmoothL1Loss(beta=0.1)  # beta 可調小一點對 heavy-tail 更穩
optimizer = torch.optim.Adam(model.parameters(), lr=5e-4, weight_decay=5e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5, verbose=False)
scaler = GradScaler()


# 批量訓練
#train_dataset = TensorDataset(X_train, Y_train)
batch_size = 32
#train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
#test_dataset = TensorDataset(X_test, Y_test)
#test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)  # 測試集分批

best_loss = float('inf')
patience = 20
counter = 0
num_epochs = 200
accumulation_steps = 4  # 梯度累積步數

for epoch in range(num_epochs):
    model.train()
    total_loss = 0
    optimizer.zero_grad()  # 在循環開始時清零梯度
    
    for i, (batch_X, batch_Y) in enumerate(train_loader):
        # non_blocking=True 在 pin_memory=True 時會更有效率
        batch_X = batch_X.to(device, non_blocking=True)
        batch_Y = batch_Y.to(device, non_blocking=True)

        batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
        with autocast():  # 混合精度
            outputs = model(batch_X).squeeze()
            loss = criterion(outputs, batch_Y) / accumulation_steps
        scaler.scale(loss).backward()
        if (i + 1) % accumulation_steps == 0:
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()
        total_loss += loss.item() * accumulation_steps

    model.eval()
    val_loss = 0
    val_outputs = []
    with torch.no_grad():
        for batch_X, batch_Y in test_loader:
            batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
            with autocast():
                outputs = model(batch_X).squeeze()
                loss = criterion(outputs, batch_Y)
            val_outputs.append(outputs)
            val_loss += loss.item()

    val_outputs = torch.cat(val_outputs)
    val_loss = val_loss / len(test_loader)
    scheduler.step(val_loss)

    if (epoch + 1) % 10 == 0:
        avg_loss = total_loss / len(train_loader)
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {avg_loss:.4f}, Val Loss: {val_loss:.4f}')

    if val_loss < best_loss:
        best_loss = val_loss
        counter = 0
        torch.save(model.state_dict(), 'best_model.pth')
    else:
        counter += 1
        if counter >= patience:
            print(f'Early stopping at epoch {epoch+1}')
            break

# 預測與評估
model.load_state_dict(torch.load('best_model.pth', weights_only=True))
model.eval()
test_predict = []
with torch.no_grad():
    for batch_X, _ in test_loader:
        batch_X = batch_X.to(device)
        with autocast():
            outputs = model(batch_X).squeeze()
        test_predict.append(outputs.cpu())

test_predict = torch.cat(test_predict)
# 計算最佳閾值
precision, recall, thresholds = precision_recall_curve(Y_test.cpu().numpy(), test_predict.numpy())
f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)
optimal_idx = np.argmax(f1_scores)
optimal_threshold = thresholds[optimal_idx]
print(f"Optimal threshold: {optimal_threshold:.4f}")

test_predict_binary = (test_predict >= optimal_threshold).float()
print(classification_report(Y_test.cpu().numpy(), test_predict_binary.numpy(), zero_division=0))
print(f'ROC AUC: {roc_auc_score(Y_test.cpu().numpy(), test_predict.numpy()):.4f}')

# 混淆矩陣
cm = confusion_matrix(Y_test.cpu().numpy(), test_predict_binary.numpy())
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

# Precision-Recall 曲線
plt.figure(figsize=(8, 6))
plt.plot(recall, precision, label='Precision-Recall Curve')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.legend()
plt.show()

# 預測概率分佈
plt.figure(figsize=(10, 6))
plt.hist(test_predict.numpy(), bins=50, alpha=0.5, label='Predicted Probabilities')
plt.axvline(x=optimal_threshold, color='r', linestyle='--', label=f'Optimal Threshold ({optimal_threshold:.4f})')
plt.xlabel('Predicted Probability')
plt.ylabel('Frequency')
plt.title('Distribution of Predicted Probabilities')
plt.legend()
plt.show()

# 保存模型和數據
torch.save(model.state_dict(), 'lstm_stock_model.pth')
with open('stock_data.pkl', 'wb') as f:
    pickle.dump((X_train, Y_train, X_test, Y_test, scalers), f)

# 清理內存
del X_train, Y_train, X_test, Y_test, val_outputs, test_predict
torch.cuda.empty_cache()
gc.collect()
#else:
#    print("No valid data found. Consider expanding the time range or stock list.")
```

```python
#2026/03/23 在ChatgPT建議下改用transformer，結果訓練時間確實減半，據說預測效果也更好，之後需要找時間上一下關於transformer相關的課程。但仍然保留LSTM的model之後還是可以比較一下預測結果的差異
###############################################################
# 最終整合版：C-mini Ultra Stable + Fast
###############################################################
import os, gc, glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from sklearn.preprocessing import MinMaxScaler
from numpy.lib.stride_tricks import sliding_window_view

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

###############################################################
# ⭐ 1. LSTM 加速設定（非常重要）
###############################################################
torch.backends.cudnn.benchmark = True
# ⚠ 不要用 torch.compile()（會導致 LSTM contiguous 錯誤）
# ⚠ 不要用 AMP（RTX A4000 FP16 無效，反而容易降低吞吐）

###############################################################
# Config
###############################################################
FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

time_step = 126
future_days = 120
chunk_size = 50  # ⭐ 分段處理 50 個 parquet，避免爆 RAM

parquet_files = glob.glob("parquet_data/*.parquet")
print(f"總 parquet：{len(parquet_files)}")

###############################################################
# ⭐ 2. Global MinMaxScaler：一次建立，不會洩漏
###############################################################
print("Step 1 - 建立全域 scaler ...")

global_min, global_max = None, None

for pq in tqdm(parquet_files):
    df = pd.read_parquet(pq, engine="pyarrow")[FEATURE_COLS].astype(np.float32)

    if global_min is None:
        global_min = df.min().to_numpy()
        global_max = df.max().to_numpy()
    else:
        global_min = np.minimum(global_min, df.min().to_numpy())
        global_max = np.maximum(global_max, df.max().to_numpy())

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))


###############################################################
# ⭐ 3. Sliding-window 分段建立 X_all / Y_all（不爆 RAM）
###############################################################
print("Step 2 - 分段 sliding window 產生所有序列 ...")

X_list = []
Y_list = []

for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start + chunk_size)
    chunk_files = parquet_files[start:end]

    feats_chunk = []
    price_chunk = []

    # 📌 讀取 chunk 內所有資料
    for pq in chunk_files:
        try:
            df = pd.read_parquet(pq, engine="pyarrow")
            if df.empty:
                continue

            drop_cols = [c for c in df.columns if c not in FEATURE_COLS + ["Date"]]
            if drop_cols:
                df.drop(columns=drop_cols, inplace=True)
            if "Date" in df.columns:
                df.drop(columns=["Date"], inplace=True)

            df = df.astype(np.float32)

            feats_chunk.append(df[FEATURE_COLS].to_numpy())
            price_chunk.append(df["Close"].to_numpy())
        except:
            continue

    if len(feats_chunk) == 0:
        continue

    feats_chunk = np.concatenate(feats_chunk)
    price_chunk = np.concatenate(price_chunk)

    # transform（一次即可）
    feats_scaled = scaler.transform(feats_chunk)

    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0:
        continue

    # ⭐ sliding window（view）
    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)  # shape: (n - time_step + 1, 126, 15)

    # 對應 y
    cur_price = price_chunk[time_step:time_step+max_i]
    fut_price = price_chunk[time_step+future_days:time_step+future_days+max_i]
    y_vals = np.log((fut_price + 1e-12) / (cur_price + 1e-12))

    # ⭐ copy 避免 stride memory 问题（很重要）
    X_list.append(windows[:max_i].copy())
    Y_list.append(y_vals.copy())

    del feats_scaled, windows, feats_chunk, price_chunk
    gc.collect()

###############################################################
# ⭐ 4. 合併全部序列（在這裡才做 concat，不會爆 RAM）
###############################################################
print("Step 3 - 合併全部序列 ...")

X_all = np.concatenate(X_list).astype(np.float32)
Y_all = np.concatenate(Y_list).astype(np.float32)

del X_list, Y_list
gc.collect()

print(f"全部序列總數：{len(Y_all):,}")

###############################################################
# ⭐ 5. Train/Test Split
###############################################################
cut = int(0.8 * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test,  Y_test)

###############################################################
# ⭐ 6. DataLoader（num_workers=0 避免複製 13GB 資料）
###############################################################
train_loader = DataLoader(
    train_ds,
    batch_size=64,
    shuffle=True,
    num_workers=0,     # ❗ 必須為 0，避免爆 RAM
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=64,
    shuffle=False,
    num_workers=0,     # ❗ 必須為 0
    pin_memory=True
)

###############################################################
# ⭐ 7. LSTM 模型（安全版本）
###############################################################
class LSTMModel(nn.Module):
    def __init__(self, input_size=len(FEATURE_COLS), hidden=128, layers=3):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size, hidden, layers,
            batch_first=True
        )
        self.fc = nn.Linear(hidden, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1])

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = LSTMModel().to(device)

# ⭐ 避免 contiguous memory 警告
model.lstm.flatten_parameters()

criterion = nn.SmoothL1Loss()
optimizer = torch.optim.Adam(model.parameters(), lr=5e-4)
    
###############################################################
# ⭐ 8. TRAIN LOOP（最穩定版）
###############################################################
num_epochs = 40
acc_steps = 2

for epoch in range(num_epochs):
    model.train()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}")

    optimizer.zero_grad(set_to_none=True)

    for i, (bx, by) in enumerate(pbar):
        bx = bx.to(device, non_blocking=True)
        by = by.to(device, non_blocking=True)

        pred = model(bx).squeeze()
        loss = criterion(pred, by) / acc_steps

        loss.backward()

        if (i + 1) % acc_steps == 0:
            optimizer.step()
            optimizer.zero_grad(set_to_none=True)

        pbar.set_postfix(loss=float(loss) * acc_steps)

    print(f"Epoch {epoch+1} 完成")

###############################################################
# ⭐ 9. EVALUATION（回歸）
###############################################################
model.eval()
y_pred = model(X_test.to(device)).detach().cpu().numpy()

print("MAE =", mean_absolute_error(Y_test, y_pred))
print("RMSE=", mean_squared_error(Y_test, y_pred, squared=False))
print("R2  =", r2_score(Y_test, y_pred))
```

```python
#2026/03/23 在ChatgPT建議下改用transformer，結果訓練時間確實減半，據說預測效果也更好，之後需要找時間上一下關於transformer相關的課程
import torch
import torch.nn as nn
# C-mini 分段處理所有 parquet → X_all, Y_all
import numpy as np
import pandas as pd
import gc
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view
from sklearn.preprocessing import MinMaxScaler
import glob
import torch
from torch.utils.data import TensorDataset, DataLoader
import torch.optim as optim
import torch.nn as nn
from tqdm import tqdm

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

time_step = 126
future_days = 120
chunk_size = 50

parquet_files = glob.glob("parquet_data/*.parquet")

# ===== Step 1: 建立全域 scaler =====
global_min, global_max = None, None

for pq in tqdm(parquet_files, desc="Global MinMax"):
    df = pd.read_parquet(pq)[FEATURE_COLS].astype(np.float32)
    if global_min is None:
        global_min = df.min().to_numpy()
        global_max = df.max().to_numpy()
    else:
        global_min = np.minimum(global_min, df.min().to_numpy())
        global_max = np.maximum(global_max, df.max().to_numpy())

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

# ===== Step 2: 分段 sliding window =====
X_list = []
Y_list = []

for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start+chunk_size)
    chunk = parquet_files[start:end]

    feats_chunk = []
    price_chunk = []

    for pq in chunk:
        df = pd.read_parquet(pq)
        if df.empty:
            continue

        cols = [c for c in df.columns if c not in FEATURE_COLS + ["Date"]]
        if cols: df.drop(columns=cols, inplace=True)
        if "Date" in df.columns: df.drop(columns=["Date"], inplace=True)

        df = df.astype(np.float32)
        feats_chunk.append(df[FEATURE_COLS].to_numpy())
        price_chunk.append(df["Close"].to_numpy())

    if len(feats_chunk)==0:
        continue

    feats_chunk = np.concatenate(feats_chunk)
    price_chunk = np.concatenate(price_chunk)

    feats_scaled = scaler.transform(feats_chunk)

    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0:
        continue

    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    cur_p = price_chunk[time_step : time_step+max_i]
    fut_p = price_chunk[time_step+future_days : time_step+future_days+max_i]

    y_vals = np.log((fut_p+1e-12)/(cur_p+1e-12))

    X_list.append(windows[:max_i].copy())
    Y_list.append(y_vals.copy())

    del feats_scaled, windows
    gc.collect()

X_all = np.concatenate(X_list).astype(np.float32)
Y_all = np.concatenate(Y_list).astype(np.float32)



cut = int(0.8 * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test,  Y_test)

train_loader = DataLoader(
    train_ds,
    batch_size=512,   # ⭐ RTX A4000 完全吃得下
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=512,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

class PatchTSTModel(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,    # ⭐ 你能用 512 batch → patch_len 放大GPU利用率
        stride=16,       # ⭐ patch overlap 50%
        d_model=256,     # ⭐ 大 model → RTX A4000 很吃得下
        n_heads=8,
        n_layers=4,
        dropout=0.1
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        # 計算 patch 數量
        # time_step = 126
        self.num_patches = (126 - patch_len) // stride + 1

        # 每個 patch flatten = patch_len × feature_dim
        patch_dim = patch_len * input_size

        # 將 patch 映射到 d_model
        self.patch_embed = nn.Linear(patch_dim, d_model)

        # Transformer Encoder
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        # 最終回歸頭
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        # x: (B, 126, 15)
        B, T, F = x.shape

        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            # 取出 patch (B, patch_len, F)
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)  # (B, num_patches, patch_dim)

        # embedding
        h = self.patch_embed(patches)          # (B, num_patches, d_model)

        # transformer
        z = self.transformer(h)

        # 取最後一個 patch embedding（最常用）
        z_last = z[:, -1, :]

        out = self.head(z_last).squeeze(-1)
        return out
    
#############################
# 4：PatchTST 的訓練迴圈（超乾淨版）
#############################

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = PatchTSTModel().to(device)
criterion = nn.SmoothL1Loss()
optimizer = optim.Adam(model.parameters(), lr=5e-4)

num_epochs = 40

for epoch in range(num_epochs):
    model.train()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for bx, by in pbar:
        bx = bx.to(device, non_blocking=True)
        by = by.to(device, non_blocking=True)

        pred = model(bx)
        loss = criterion(pred, by)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        pbar.set_postfix(loss=float(loss))
        

#############################
# 評估（回歸）
#############################
        
model.eval()

with torch.no_grad():
    pred = model(X_test.to(device)).cpu().numpy()

print("MAE =", mean_absolute_error(Y_test, pred))
print("RMSE=", mean_squared_error(Y_test, pred, squared=False))
print("R2  =", r2_score(Y_test, pred))
```

```python
#2026.5.3 修正Global min max 變成NaN的問題後，模型終於開始有學點東西了
import os
import gc
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view
from sklearn.metrics import roc_curve, auc
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    precision_recall_curve, classification_report,
    confusion_matrix, roc_auc_score
)
import matplotlib.pyplot as plt


# =========================================================
# Config
# =========================================================
FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

TIME_STEP = 126
FUTURE_DAYS = 120
WINDOW_STRIDE = 1        # ✅ 每天滑動 = 1；若想降記憶體/減少重疊可改成 5
TRAIN_RATIO = 0.8
BATCH_SIZE = 512
NUM_WORKERS = 0
PARQUET_GLOB = "parquet_data/*.parquet"

RET_TH = np.log(1.5)     # +50% 門檻，用於 weighted regression & classification eval
ALPHA = 5.0              # 正類權重倍數（weighted regression）
NUM_EPOCHS = 40
LR = 5e-4

parquet_files = glob.glob(PARQUET_GLOB)
print(f"Found parquet files: {len(parquet_files)}")


# =========================================================
# Safe parquet reader
# =========================================================
def safe_read_parquet_features(pq, feature_cols, keep_date=False):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺欄位 -> None
    - 將 NaN / inf 清理掉
    """
    try:
        df = pd.read_parquet(pq)

        # 1) 空檔
        if df is None or df.empty:
            return None

        # 2) 檢查必要欄位
        missing_cols = [c for c in feature_cols if c not in df.columns]
        if len(missing_cols) > 0:
            print(f"[SKIP missing cols] {os.path.basename(pq)} -> {missing_cols}")
            return None

        # 3) 保留需要欄位
        keep_cols = list(feature_cols)
        if keep_date and "Date" in df.columns:
            keep_cols.append("Date")

        df = df[keep_cols].copy()

        # 4) 數值化
        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)

        # 5) 清理 NaN / inf
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty or len(df) == 0:
            return None

        return df

    except Exception as e:
        print(f"[SKIP read error] {os.path.basename(pq)} -> {e}")
        return None


# =========================================================
# Step 1. Build robust global scaler
# =========================================================
global_min, global_max = None, None
valid_files_step1 = 0
skipped_files_step1 = 0

for pq in tqdm(parquet_files, desc="Step1 Global MinMax"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step1 += 1
        continue

    arr = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

    mn = np.nanmin(arr, axis=0)
    mx = np.nanmax(arr, axis=0)

    if not np.isfinite(mn).all() or not np.isfinite(mx).all():
        print(f"[SKIP invalid min/max] {os.path.basename(pq)}")
        skipped_files_step1 += 1
        continue

    if global_min is None:
        global_min = mn
        global_max = mx
    else:
        global_min = np.minimum(global_min, mn)
        global_max = np.maximum(global_max, mx)

    valid_files_step1 += 1

if valid_files_step1 == 0:
    raise ValueError("Step 1 沒有任何有效 parquet 可建立 scaler，請檢查資料")

print("\n=== Step 1 summary ===")
print("valid_files_step1   :", valid_files_step1)
print("skipped_files_step1 :", skipped_files_step1)

global_debug_df = pd.DataFrame({
    "feature": FEATURE_COLS,
    "global_min": global_min,
    "global_max": global_max,
    "global_range": global_max - global_min
})
print(global_debug_df)

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

assert np.isfinite(scaler.data_min_).all()
assert np.isfinite(scaler.data_max_).all()
assert np.isfinite(scaler.data_range_).all()

print("\n=== sklearn scaler internal stats ===")
print("data_min_ :", scaler.data_min_)
print("data_max_ :", scaler.data_max_)
print("data_range_:", scaler.data_range_)


# =========================================================
# Step 2. Build windows (per-stock, no cross-stock contamination)
# =========================================================
X_list = []
Y_list = []

valid_files_step2 = 0
skipped_files_step2 = 0

print("\n========= Step 2: sliding windows =========")
print(f"TIME_STEP={TIME_STEP}, FUTURE_DAYS={FUTURE_DAYS}, WINDOW_STRIDE={WINDOW_STRIDE}")

for pq in tqdm(parquet_files, desc="Step2 Build Windows"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step2 += 1
        continue

    feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    close_raw = df["Close"].to_numpy(dtype=np.float32)

    if len(feats) == 0 or len(close_raw) == 0:
        skipped_files_step2 += 1
        continue

    feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)

    feats_scaled = scaler.transform(feats)
    feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

    n = len(feats_scaled)
    max_i = n - TIME_STEP - FUTURE_DAYS
    if max_i <= 0:
        skipped_files_step2 += 1
        continue

    # 每支股票各自做 sliding window（避免跨股票污染）
    windows = sliding_window_view(
        feats_scaled,
        (TIME_STEP, feats_scaled.shape[1])
    ).squeeze(1)

    # 只取有效範圍
    windows = windows[:max_i]

    # stride 下採樣
    windows = windows[::WINDOW_STRIDE]

    cur_p = close_raw[TIME_STEP : TIME_STEP + max_i : WINDOW_STRIDE]
    fut_p = close_raw[TIME_STEP + FUTURE_DAYS : TIME_STEP + FUTURE_DAYS + max_i : WINDOW_STRIDE]

    y_vals = np.log((fut_p + 1e-12) / (cur_p + 1e-12))
    y_vals = np.nan_to_num(y_vals, nan=0.0, posinf=0.0, neginf=0.0)

    if len(windows) == 0 or len(y_vals) == 0:
        skipped_files_step2 += 1
        continue

    X_list.append(windows.astype(np.float32, copy=True))
    Y_list.append(y_vals.astype(np.float32, copy=True))

    valid_files_step2 += 1

    del feats, feats_scaled, windows, close_raw, cur_p, fut_p, y_vals
    gc.collect()

if len(X_list) == 0 or len(Y_list) == 0:
    raise ValueError("Step 2 沒有產生任何有效樣本，請檢查 parquet 資料與參數")

print("\n=== Step 2 summary ===")
print("valid_files_step2   :", valid_files_step2)
print("skipped_files_step2 :", skipped_files_step2)

X_all = np.concatenate(X_list, axis=0).astype(np.float32)
Y_all = np.concatenate(Y_list, axis=0).astype(np.float32)

print("X_all shape:", X_all.shape)
print("Y_all shape:", Y_all.shape)
print("X_all NaN? ", np.isnan(X_all).any())
print("Y_all NaN? ", np.isnan(Y_all).any())
print("X_all min/max/std:", X_all.min(), X_all.max(), X_all.std())
print("Y_all min/max/std:", Y_all.min(), Y_all.max(), Y_all.std())


# =========================================================
# Step 3. Train / Test split
# =========================================================
cut = int(TRAIN_RATIO * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

print("\n=== Train / Test ===")
print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
print("X_test shape :", X_test.shape)
print("Y_test shape :", Y_test.shape)

assert X_train.ndim == 3 and X_train.shape[1] == TIME_STEP and X_train.shape[2] == len(FEATURE_COLS)
assert X_test.ndim == 3 and X_test.shape[1] == TIME_STEP and X_test.shape[2] == len(FEATURE_COLS)


# =========================================================
# DataLoader
# =========================================================
train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test, Y_test)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)


# =========================================================
# PatchTST Model
# =========================================================
class PatchTSTModel(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,
        stride=16,
        d_model=128,
        n_heads=4,
        n_layers=3,
        dropout=0.1
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        patch_dim = patch_len * input_size
        self.patch_embed = nn.Linear(patch_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        B, T, F = x.shape

        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)
        h = self.patch_embed(patches)
        z = self.transformer(h)
        z_last = z[:, -1, :]
        out = self.head(z_last).squeeze(-1)
        return out


# =========================================================
# Weighted regression loss
# =========================================================
def weighted_regression_loss(pred, target, alpha=5.0, ret_th=np.log(1.5)):
    """
    對 >= ret_th 的正類樣本提高 loss 權重
    """
    base_loss = F.smooth_l1_loss(pred, target, reduction='none')
    weights = torch.ones_like(target)
    weights[target >= ret_th] = alpha
    return (weights * base_loss).mean()


# =========================================================
# Train
# =========================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("\nDevice:", device)

model = PatchTSTModel(input_size=len(FEATURE_COLS)).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR)

for epoch in range(NUM_EPOCHS):
    model.train()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")

    for bx, by in pbar:
        bx = bx.to(device, non_blocking=True)
        by = by.to(device, non_blocking=True)

        pred = model(bx)
        loss = weighted_regression_loss(pred, by, alpha=ALPHA, ret_th=RET_TH)

        optimizer.zero_grad()
        loss.backward()

        # 防梯度爆炸
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

        optimizer.step()
        pbar.set_postfix(loss=float(loss))

    print(f"Epoch {epoch+1} 完成")


# =========================================================
# Evaluation (batch inference, avoid OOM)
# =========================================================
model.eval()
preds = []
trues = []

with torch.no_grad():
    for bx, by in DataLoader(test_ds, batch_size=BATCH_SIZE, num_workers=0):
        bx = bx.to(device, non_blocking=True)
        pred = model(bx).cpu()
        preds.append(pred)
        trues.append(by)

pred = torch.cat(preds).numpy()
true = torch.cat(trues).numpy()

print("\n=== Eval sanity check ===")
print("pred NaN?", np.isnan(pred).any())
print("true NaN?", np.isnan(true).any())

# Regression metrics（可選）
print("\n=== Regression Metrics ===")
print("MAE  =", mean_absolute_error(true, pred))
print("RMSE =", mean_squared_error(true, pred, squared=False))
print("R²   =", r2_score(true, pred))

# Classification by threshold on regression score
y_true_cls = (true >= RET_TH).astype(int)

precision, recall, thresholds = precision_recall_curve(y_true_cls, pred)

f1_scores = 2 * precision[:-1] * recall[:-1] / (precision[:-1] + recall[:-1] + 1e-12)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print("\n=== Classification Metrics ===")
print("Best threshold (on model score):", best_threshold)
print("Best F1:", f1_scores[best_idx])

y_pred_cls = (pred >= best_threshold).astype(int)

print(classification_report(
    y_true_cls,
    y_pred_cls,
    target_names=["Not >=50%", ">=50%"],
    digits=4,
    zero_division=0
))

cm = confusion_matrix(y_true_cls, y_pred_cls)
print("Confusion Matrix:\n", cm)

try:
    auc_score = roc_auc_score(y_true_cls, pred)
    print("ROC-AUC:", auc_score)
except Exception as e:
    print("ROC-AUC failed:", e)


# =========================================================
# Plot PR / ROC
# =========================================================
plt.figure(figsize=(6,5))
plt.plot(recall, precision, label="PR curve")
plt.scatter(recall[best_idx], precision[best_idx], color="red", label="Best F1")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid(True)
plt.show()

fpr, tpr, _ = roc_curve(y_true_cls, pred)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.4f})")
plt.plot([0,1], [0,1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#2026.04.14 這個是Multi‑task：Regression + Classification的版本
#2026.5.10 修正global min max可能會有NaN的問題，但這個model成效很差，AI說用classification一刀切加上正負資料不平均會造成效果很差
import os
import gc
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    precision_recall_curve, classification_report,
    confusion_matrix, roc_auc_score
)
import matplotlib.pyplot as plt


# =========================================================
# Config
# =========================================================
FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

TIME_STEP = 126
FUTURE_DAYS = 120
WINDOW_STRIDE = 1        # ✅ 每天滑動 = 1；若想降記憶體/減少重疊可改成 5
TRAIN_RATIO = 0.8
BATCH_SIZE = 512
NUM_WORKERS = 0
PARQUET_GLOB = "parquet_data/*.parquet"

RET_TH = np.log(1.5)     # +50% 門檻，用於 weighted regression & classification eval
ALPHA = 5.0              # 正類權重倍數（weighted regression）
NUM_EPOCHS = 40
LR = 5e-4

parquet_files = glob.glob(PARQUET_GLOB)
print(f"Found parquet files: {len(parquet_files)}")


# =========================================================
# Safe parquet reader
# =========================================================
def safe_read_parquet_features(pq, feature_cols, keep_date=False):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺欄位 -> None
    - 將 NaN / inf 清理掉
    """
    try:
        df = pd.read_parquet(pq)

        # 1) 空檔
        if df is None or df.empty:
            return None

        # 2) 檢查必要欄位
        missing_cols = [c for c in feature_cols if c not in df.columns]
        if len(missing_cols) > 0:
            print(f"[SKIP missing cols] {os.path.basename(pq)} -> {missing_cols}")
            return None

        # 3) 保留需要欄位
        keep_cols = list(feature_cols)
        if keep_date and "Date" in df.columns:
            keep_cols.append("Date")

        df = df[keep_cols].copy()

        # 4) 數值化
        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)

        # 5) 清理 NaN / inf
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty or len(df) == 0:
            return None

        return df

    except Exception as e:
        print(f"[SKIP read error] {os.path.basename(pq)} -> {e}")
        return None


# =========================================================
# Step 1. Build robust global scaler
# =========================================================
global_min, global_max = None, None
valid_files_step1 = 0
skipped_files_step1 = 0

for pq in tqdm(parquet_files, desc="Step1 Global MinMax"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step1 += 1
        continue

    arr = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

    mn = np.nanmin(arr, axis=0)
    mx = np.nanmax(arr, axis=0)

    if not np.isfinite(mn).all() or not np.isfinite(mx).all():
        print(f"[SKIP invalid min/max] {os.path.basename(pq)}")
        skipped_files_step1 += 1
        continue

    if global_min is None:
        global_min = mn
        global_max = mx
    else:
        global_min = np.minimum(global_min, mn)
        global_max = np.maximum(global_max, mx)

    valid_files_step1 += 1

if valid_files_step1 == 0:
    raise ValueError("Step 1 沒有任何有效 parquet 可建立 scaler，請檢查資料")

print("\n=== Step 1 summary ===")
print("valid_files_step1   :", valid_files_step1)
print("skipped_files_step1 :", skipped_files_step1)

global_debug_df = pd.DataFrame({
    "feature": FEATURE_COLS,
    "global_min": global_min,
    "global_max": global_max,
    "global_range": global_max - global_min
})
print(global_debug_df)

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

assert np.isfinite(scaler.data_min_).all()
assert np.isfinite(scaler.data_max_).all()
assert np.isfinite(scaler.data_range_).all()

print("\n=== sklearn scaler internal stats ===")
print("data_min_ :", scaler.data_min_)
print("data_max_ :", scaler.data_max_)
print("data_range_:", scaler.data_range_)


# =========================================================
# Step 2. Build windows (per-stock, no cross-stock contamination)
# =========================================================
X_list = []
Y_list = []

valid_files_step2 = 0
skipped_files_step2 = 0

print("\n========= Step 2: sliding windows =========")
print(f"TIME_STEP={TIME_STEP}, FUTURE_DAYS={FUTURE_DAYS}, WINDOW_STRIDE={WINDOW_STRIDE}")

for pq in tqdm(parquet_files, desc="Step2 Build Windows"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step2 += 1
        continue

    feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    close_raw = df["Close"].to_numpy(dtype=np.float32)

    if len(feats) == 0 or len(close_raw) == 0:
        skipped_files_step2 += 1
        continue

    feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)

    feats_scaled = scaler.transform(feats)
    feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

    n = len(feats_scaled)
    max_i = n - TIME_STEP - FUTURE_DAYS
    if max_i <= 0:
        skipped_files_step2 += 1
        continue

    # 每支股票各自做 sliding window（避免跨股票污染）
    windows = sliding_window_view(
        feats_scaled,
        (TIME_STEP, feats_scaled.shape[1])
    ).squeeze(1)

    # 只取有效範圍
    windows = windows[:max_i]

    # stride 下採樣
    windows = windows[::WINDOW_STRIDE]

    cur_p = close_raw[TIME_STEP : TIME_STEP + max_i : WINDOW_STRIDE]
    fut_p = close_raw[TIME_STEP + FUTURE_DAYS : TIME_STEP + FUTURE_DAYS + max_i : WINDOW_STRIDE]

    y_vals = np.log((fut_p + 1e-12) / (cur_p + 1e-12))
    y_vals = np.nan_to_num(y_vals, nan=0.0, posinf=0.0, neginf=0.0)

    if len(windows) == 0 or len(y_vals) == 0:
        skipped_files_step2 += 1
        continue

    X_list.append(windows.astype(np.float32, copy=True))
    Y_list.append(y_vals.astype(np.float32, copy=True))

    valid_files_step2 += 1

    del feats, feats_scaled, windows, close_raw, cur_p, fut_p, y_vals
    gc.collect()

if len(X_list) == 0 or len(Y_list) == 0:
    raise ValueError("Step 2 沒有產生任何有效樣本，請檢查 parquet 資料與參數")

print("\n=== Step 2 summary ===")
print("valid_files_step2   :", valid_files_step2)
print("skipped_files_step2 :", skipped_files_step2)

X_all = np.concatenate(X_list, axis=0).astype(np.float32)
Y_all = np.concatenate(Y_list, axis=0).astype(np.float32)

print("X_all shape:", X_all.shape)
print("Y_all shape:", Y_all.shape)
print("X_all NaN? ", np.isnan(X_all).any())
print("Y_all NaN? ", np.isnan(Y_all).any())
print("X_all min/max/std:", X_all.min(), X_all.max(), X_all.std())
print("Y_all min/max/std:", Y_all.min(), Y_all.max(), Y_all.std())

cut = int(0.8 * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test,  Y_test)

train_loader = DataLoader(
    train_ds,
    batch_size=512,  # ⭐ RTX A4000 可承受
    shuffle=True,
    num_workers=0,   # ⭐ 避免 RAM duplicating
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=512,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)



class PatchTSTMultiTask(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,
        stride=16,
        d_model=128,
        n_heads=4,
        n_layers=3,
        dropout=0.1
    ):
        super().__init__()
        self.patch_len = patch_len
        self.stride = stride

        patch_dim = patch_len * input_size
        self.patch_embed = nn.Linear(patch_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(
            encoder_layer, n_layers
        )

        # —— Regression head（連續報酬）
        self.reg_head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1),
        )

        # —— Classification head（是否 >=50%）
        self.cls_head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1),   # logits
        )

    def forward(self, x):
        B, T, F = x.shape
        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)
        h = self.patch_embed(patches)
        z = self.transformer(h)

        z_last = z[:, -1, :]

        reg_out = self.reg_head(z_last).squeeze(-1)   # regression
        cls_logit = self.cls_head(z_last).squeeze(-1) # classification (logits)

        return reg_out, cls_logit



device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
#pos_weight = torch.tensor([(1-0.08)/0.08]).to(device)


RET_TH = np.log(1.3)
y_train_cls = (Y_train.numpy() >= RET_TH).astype(np.int32)
pos = y_train_cls.sum()
neg = len(y_train_cls) - pos
pos_weight = torch.tensor([neg / max(pos, 1)], device=device, dtype=torch.float32)
print("pos_rate(train)=", pos/len(y_train_cls), "pos_weight=", pos_weight.item())


model = PatchTSTMultiTask().to(device)

reg_criterion = nn.SmoothL1Loss()
cls_criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

optimizer = torch.optim.Adam(model.parameters(), lr=5e-4)

λ = 3  # classification loss weight（可之後調 0.5~2）

num_epochs = 20

for epoch in range(num_epochs):
    model.train()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for bx, by in pbar:
        bx = bx.to(device)
        by = by.to(device)

        # —— Label for classification
        y_cls = (by >= RET_TH).float()

        reg_pred, cls_logit = model(bx)

        loss_reg = reg_criterion(reg_pred, by)
        loss_cls = cls_criterion(cls_logit, y_cls)

        loss = loss_reg + λ * loss_cls

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        
        
        with torch.no_grad():
            prob = torch.sigmoid(cls_logit)
            pbar.set_postfix(
                loss=float(loss),
                reg=float(loss_reg),
                cls=float(loss_cls),
                prob_mean=float(prob.mean()),
                prob_p90=float(torch.quantile(prob, 0.9))
            )


        #pbar.set_postfix(
        #    loss=float(loss),
        #    reg=float(loss_reg),
        #    cls=float(loss_cls)
        #)

    print(f"Epoch {epoch+1} 完成")


from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    precision_recall_curve, classification_report, confusion_matrix, roc_auc_score
)
    

model.eval()
cls_scores = []
true_cls = []

with torch.no_grad():
    
    for bx, by in test_loader:
        bx = bx.to(device)
        by = by.to(device)

        _, cls_logit = model(bx)
        #cls_scores.append(torch.sigmoid(cls_logit).cpu())
        cls_scores.append(cls_logit.cpu())
        true_cls.append((by >= RET_TH).float().cpu())

cls_scores = torch.cat(cls_scores).numpy()
true_cls = torch.cat(true_cls).numpy()

from sklearn.metrics import average_precision_score, roc_auc_score, classification_report, confusion_matrix
print("score std:", cls_scores.std())
print("score quantiles:", np.quantile(cls_scores, [0, 0.01, 0.1, 0.5, 0.9, 0.99, 1.0]))
print("AP:", average_precision_score(true_cls, cls_scores))
print("ROC-AUC:", roc_auc_score(true_cls, cls_scores))
print("Base rate:", true_cls.mean())

# ---- Top-k 試法（例如 2%）----
k = 0.02
thr = np.quantile(cls_scores, 1-k)
y_pred_topk = (cls_scores >= thr).astype(int)

print("Top-k threshold:", thr)
print("Predicted positive rate:", y_pred_topk.mean())
print(confusion_matrix(true_cls, y_pred_topk))
print(classification_report(true_cls, y_pred_topk, zero_division=0))


# 30% return 的 log 門檻
RET_TH = np.log(1.3)  # ≈ 0.405465


# ⭐ 再次 NaN 檢查
print("cls_scores NaN?", np.isnan(cls_scores).any())
print("true_cls NaN?", np.isnan(true_cls).any())

from sklearn.metrics import precision_recall_curve
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

precision, recall, thresholds = precision_recall_curve(
    true_cls, cls_scores
)

# F1 score
f1_scores = 2 * precision[:-1] * recall[:-1] / (precision[:-1] + recall[:-1] + 1e-12)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print("Best threshold (on model score):", best_threshold)
print("Best F1:", f1_scores[best_idx])



y_pred_cls = (cls_scores >= best_threshold).astype(int)

print(classification_report(
    true_cls,
    y_pred_cls,
    target_names=["Not >=50%", ">=50%"],
    digits=4
))

cm = confusion_matrix(true_cls, y_pred_cls)
print("Confusion Matrix:\n", cm)

#print("MAE  =", mean_absolute_error(true, pred))
#print("RMSE =", mean_squared_error(true, pred, squared=False))
#print("R²    =", r2_score(true, pred))



# ------- PR Curve -------
plt.figure(figsize=(6,5))
plt.plot(recall, precision, label="PR curve")
plt.scatter(
    recall[best_idx],
    precision[best_idx],
    color="red",
    label="Best F1"
)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid(True)
plt.show()

# ------- ROC Curve -------
fpr, tpr, _ = roc_curve(true_cls, cls_scores)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.4f})")
plt.plot([0,1], [0,1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid(True)
plt.show()

```

```python
import os
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt

# =========================
# 設定
# =========================
PARQUET_GLOB = "parquet_data/*.parquet"

time_step = 126
future_days = 120

# 事件門檻（log-return）
RET_TH_30 = np.log(1.3)   # +30%
RET_TH_50 = np.log(1.5)   # +50%

# 是否要儲存 CSV
SAVE_CSV = True
CSV_PATH = "stock_positive_rate_summary.csv"


# =========================
# 單檔分析函式
# =========================
def analyze_one_stock(parquet_path, time_step=126, future_days=120,
                      ret_th_30=np.log(1.3), ret_th_50=np.log(1.5)):
    """
    回傳單一股票的事件統計結果(dict)
    """
    stock_code = os.path.basename(parquet_path).replace(".parquet", "")

    try:
        df = pd.read_parquet(parquet_path)

        if df.empty:
            return None

        # 只保留必要欄位
        keep_cols = ["Close"]
        if "Date" in df.columns:
            keep_cols.append("Date")

        missing_cols = [c for c in keep_cols if c not in df.columns]
        if missing_cols:
            print(f"[WARN] {stock_code} 缺少欄位: {missing_cols}")
            return None

        df = df[keep_cols].copy()

        # 日期排序（如果有 Date）
        if "Date" in df.columns:
            df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
            df = df.sort_values("Date").reset_index(drop=True)

        # Close 修補
        df["Close"] = pd.to_numeric(df["Close"], errors="coerce")
        df["Close"] = df["Close"].ffill().bfill()

        # 如果 Close 還有 NaN 就跳過
        if df["Close"].isna().any():
            print(f"[WARN] {stock_code} Close 仍有 NaN，跳過")
            return None

        close_raw = df["Close"].to_numpy(dtype=np.float32)
        n_rows = len(close_raw)

        total_windows = n_rows - time_step - future_days
        if total_windows <= 0:
            return None

        # 建立 label（log return）
        current_price = close_raw[time_step : time_step + total_windows]
        future_price  = close_raw[time_step + future_days : time_step + future_days + total_windows]

        log_ret = np.log((future_price + 1e-12) / (current_price + 1e-12))

        pos_30 = (log_ret >= ret_th_30)
        pos_50 = (log_ret >= ret_th_50)

        result = {
            "stock_code": stock_code,
            "n_rows": n_rows,
            "total_windows": total_windows,
            "pos_30_count": int(pos_30.sum()),
            "pos_50_count": int(pos_50.sum()),
            "pos_30_rate": float(pos_30.mean()),
            "pos_50_rate": float(pos_50.mean()),
            "close_mean": float(np.mean(close_raw)),
            "close_std": float(np.std(close_raw)),
        }

        if "Date" in df.columns:
            result["start_date"] = df["Date"].min()
            result["end_date"] = df["Date"].max()
        else:
            result["start_date"] = pd.NaT
            result["end_date"] = pd.NaT

        return result

    except Exception as e:
        print(f"[ERROR] {stock_code}: {e}")
        return None


# =========================
# 全部股票分析
# =========================
parquet_files = glob.glob(PARQUET_GLOB)
print(f"找到 parquet 數量: {len(parquet_files)}")

rows = []

for pq in tqdm(parquet_files, desc="Analyzing stocks"):
    r = analyze_one_stock(
        pq,
        time_step=time_step,
        future_days=future_days,
        ret_th_30=RET_TH_30,
        ret_th_50=RET_TH_50
    )
    if r is not None:
        rows.append(r)

summary_df = pd.DataFrame(rows)

if summary_df.empty:
    raise ValueError("沒有任何有效股票資料可供統計")

# 排序：先看 50% 事件比例
summary_df = summary_df.sort_values(
    by=["pos_50_rate", "pos_30_rate", "total_windows"],
    ascending=[False, False, False]
).reset_index(drop=True)

# =========================
# 全市場統計
# =========================
total_windows_all = summary_df["total_windows"].sum()
total_pos_30_all = summary_df["pos_30_count"].sum()
total_pos_50_all = summary_df["pos_50_count"].sum()

weighted_pos_30_rate = total_pos_30_all / total_windows_all
weighted_pos_50_rate = total_pos_50_all / total_windows_all

mean_stock_pos_30_rate = summary_df["pos_30_rate"].mean()
mean_stock_pos_50_rate = summary_df["pos_50_rate"].mean()

median_stock_pos_30_rate = summary_df["pos_30_rate"].median()
median_stock_pos_50_rate = summary_df["pos_50_rate"].median()

print("\n================ 全市場摘要 ================")
print(f"有效股票數量                : {len(summary_df)}")
print(f"總樣本數 total_windows      : {total_windows_all:,}")
print(f"30% 正類總數                : {total_pos_30_all:,}")
print(f"50% 正類總數                : {total_pos_50_all:,}")
print(f"30% 加權平均正類比例         : {weighted_pos_30_rate:.6f}")
print(f"50% 加權平均正類比例         : {weighted_pos_50_rate:.6f}")
print(f"30% 每股平均正類比例         : {mean_stock_pos_30_rate:.6f}")
print(f"50% 每股平均正類比例         : {mean_stock_pos_50_rate:.6f}")
print(f"30% 每股中位數正類比例       : {median_stock_pos_30_rate:.6f}")
print(f"50% 每股中位數正類比例       : {median_stock_pos_50_rate:.6f}")

# =========================
# Top / Bottom 股票
# =========================
print("\n================ Top 20：50% 正類比例最高的股票 ================")
print(summary_df[[
    "stock_code", "total_windows", "pos_50_count", "pos_50_rate",
    "pos_30_count", "pos_30_rate"
]].head(20).to_string(index=False))

print("\n================ Bottom 20：50% 正類比例最低的股票 ================")
print(summary_df[[
    "stock_code", "total_windows", "pos_50_count", "pos_50_rate",
    "pos_30_count", "pos_30_rate"
]].tail(20).to_string(index=False))

# =========================
# 儲存 CSV
# =========================
if SAVE_CSV:
    summary_df.to_csv(CSV_PATH, index=False, encoding="utf-8-sig")
    print(f"\n已儲存 CSV：{CSV_PATH}")

# =========================
# 圖表：50% / 30% 正類比例分布
# =========================
plt.figure(figsize=(10, 5))
plt.hist(summary_df["pos_50_rate"], bins=50, alpha=0.7)
plt.xlabel("Per-stock positive rate (>=50%)")
plt.ylabel("Count of stocks")
plt.title("Distribution of per-stock >=50% positive rate")
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.hist(summary_df["pos_30_rate"], bins=50, alpha=0.7)
plt.xlabel("Per-stock positive rate (>=30%)")
plt.ylabel("Count of stocks")
plt.title("Distribution of per-stock >=30% positive rate")
plt.grid(True)
plt.show()

# =========================
# 圖表：Top 20 股票（50%）
# =========================
top20 = summary_df.head(20).copy()

plt.figure(figsize=(12, 6))
plt.bar(top20["stock_code"], top20["pos_50_rate"])
plt.xticks(rotation=90)
plt.ylabel("Positive rate (>=50%)")
plt.title("Top 20 stocks by >=50% positive rate")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```python
#Autoencoder版本降維
import os
import gc
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm


import pandas as pd
from numpy.lib.stride_tricks import sliding_window_view
from sklearn.preprocessing import MinMaxScaler
import glob
import torch.nn.functional as F

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

time_step = 126
future_days = 120
chunk_size = 50

parquet_files = glob.glob("parquet_data/*.parquet")

# ========= Step 1: 全域 scaler =========
global_min, global_max = None, None
for pq in tqdm(parquet_files, desc="Global MinMax"):
    df = pd.read_parquet(pq)[FEATURE_COLS].astype(np.float32)
    
    # ⭐ NaN 修補（避免全 NaN）
    df = df.ffill().bfill().fillna(0)
    if df.empty:
        continue

    if global_min is None:
        global_min = df.min().to_numpy()
        global_max = df.max().to_numpy()
    else:
        global_min = np.minimum(global_min, df.min().to_numpy())
        global_max = np.maximum(global_max, df.max().to_numpy())

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

# ========= Step 2: 分段 sliding window (C-mini) =========
X_list = []
Y_list = []
print("========= Step 2: 分段 sliding window (C-mini) =========")
for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start + chunk_size)
    chunk_files = parquet_files[start:end]

    feats_chunk = []
    price_chunk = []

    for pq in chunk_files:
        df = pd.read_parquet(pq)
        if df.empty: continue

        drop_cols = [c for c in df.columns if c not in FEATURE_COLS + ["Date"]]
        if drop_cols: df.drop(columns=drop_cols, inplace=True)
        if "Date" in df.columns: df.drop(columns=["Date"], inplace=True)

        df = df.astype(np.float32)
        
        # ⭐ NaN 修補（來源層）
        df = df.ffill().bfill().fillna(0)

        feats_chunk.append(df[FEATURE_COLS].to_numpy())
        price_chunk.append(df["Close"].to_numpy())

    if len(feats_chunk) == 0: continue

    feats_chunk = np.concatenate(feats_chunk)
    price_chunk = np.concatenate(price_chunk)
    
    
    # ⭐ 修補 NaN
    feats_chunk = np.nan_to_num(feats_chunk)

    
    feats_scaled = scaler.transform(feats_chunk)
    
    # ⭐ 修補 scaler 後 NaN
    feats_scaled = np.nan_to_num(feats_scaled)

    
    # 只在真的有 NaN/inf 時才補（避免每次都拷貝）
    if not np.isfinite(feats_scaled).all():
        feats_scaled = np.nan_to_num(feats_scaled)

    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0: continue
        
    windows = sliding_window_view(
        feats_scaled, (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    
    cur_p = price_chunk[time_step : time_step+max_i]
    fut_p = price_chunk[time_step+future_days : time_step+future_days+max_i]
    y_vals = np.log((fut_p + 1e-12) / (cur_p + 1e-12))
    y_vals = np.nan_to_num(y_vals)

    X_list.append(windows[:max_i])
    Y_list.append(y_vals)

    del feats_scaled, windows
    gc.collect()

X_all = np.concatenate(X_list).astype(np.float32)
Y_all = np.concatenate(Y_list).astype(np.float32)


# ✅ 全域檢查 NaN（保證沒有）
print("X_all NaN?", np.isnan(X_all).any())
print("Y_all NaN?", np.isnan(Y_all).any())

cut = int(0.8 * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

assert X_train.ndim == 3 and X_train.shape[1] == 126 and X_train.shape[2] == 14
assert X_test.ndim == 3 and X_test.shape[1] == 126 and X_test.shape[2] == 14


batch_size = 512   # A4000 通常可先試 512；如果 OOM 就降到 256
num_workers = 0    # 你的資料很大，先用 0 比較穩

train_ds = TensorDataset(X_train)
test_ds  = TensorDataset(X_test)

train_loader = DataLoader(
    train_ds,
    batch_size=batch_size,
    shuffle=True,
    num_workers=num_workers,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=batch_size,
    shuffle=False,
    num_workers=num_workers,
    pin_memory=True
)

class ConvSeqAutoencoder(nn.Module):
    def __init__(self, input_dim=14, seq_len=126, latent_dim=64):
        super().__init__()
        self.input_dim = input_dim
        self.seq_len = seq_len
        self.latent_dim = latent_dim

        # -------- Encoder --------
        # 輸入 shape: (B, 14, 126)
        self.encoder = nn.Sequential(
            nn.Conv1d(input_dim, 32, kernel_size=5, stride=2, padding=2),   # 126 -> 63
            nn.ReLU(),
            nn.Conv1d(32, 64, kernel_size=5, stride=2, padding=2),          # 63 -> 32
            nn.ReLU(),
            nn.Conv1d(64, 128, kernel_size=5, stride=2, padding=2),         # 32 -> 16
            nn.ReLU(),
        )

        # 最後會是 (B, 128, 16)
        self.flatten_dim = 128 * 16
        self.to_latent = nn.Linear(self.flatten_dim, latent_dim)

        # -------- Decoder --------
        self.from_latent = nn.Linear(latent_dim, self.flatten_dim)

        self.decoder = nn.Sequential(
            # (B, 128, 16)
            nn.ConvTranspose1d(128, 64, kernel_size=4, stride=2, padding=1),           # 16 -> 32
            nn.ReLU(),
            nn.ConvTranspose1d(64, 32, kernel_size=5, stride=2, padding=2),            # 32 -> 63
            nn.ReLU(),
            nn.ConvTranspose1d(32, input_dim, kernel_size=4, stride=2, padding=1),     # 63 -> 126
        )

    def encode(self, x):
        """
        x: (B, 126, 14)
        return: z -> (B, latent_dim)
        """
        x = x.transpose(1, 2)          # -> (B, 14, 126)
        h = self.encoder(x)            # -> (B, 128, 16)
        h = h.reshape(h.size(0), -1)   # -> (B, 2048)
        z = self.to_latent(h)          # -> (B, latent_dim)
        return z

    def decode(self, z):
        """
        z: (B, latent_dim)
        return x_hat: (B, 126, 14)
        """
        h = self.from_latent(z)        # -> (B, 2048)
        h = h.view(h.size(0), 128, 16) # -> (B, 128, 16)
        x_hat = self.decoder(h)        # -> (B, 14, 126)
        x_hat = x_hat.transpose(1, 2)  # -> (B, 126, 14)
        return x_hat

    def forward(self, x):
        z = self.encode(x)
        x_hat = self.decode(z)
        return x_hat, z
    
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

model = ConvSeqAutoencoder(
    input_dim=14,
    seq_len=126,
    latent_dim=64   # 你可改成 32 / 64 / 128
).to(device)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)


num_epochs = 30
best_val_loss = float("inf")
patience = 3
wait = 0
min_delta = 1e-7   # 改善小於這個就不算有進步

save_path = "best_autoencoder.pth"

for epoch in range(num_epochs):
    # ===== Train =====
    model.train()
    train_loss_sum = 0.0
    train_count = 0

    pbar = tqdm(train_loader, desc=f"AE Epoch {epoch+1}/{num_epochs}")

    for (bx,) in pbar:
        bx = bx.to(device, non_blocking=True)

        x_hat, z = model(bx)
        loss = criterion(x_hat, bx)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss_sum += loss.item() * bx.size(0)
        train_count += bx.size(0)

        pbar.set_postfix(loss=float(loss))

    train_loss = train_loss_sum / max(train_count, 1)

    # ===== Validation =====
    model.eval()
    val_loss_sum = 0.0
    val_count = 0

    with torch.no_grad():
        for (bx,) in test_loader:
            bx = bx.to(device, non_blocking=True)
            x_hat, z = model(bx)
            loss = criterion(x_hat, bx)

            val_loss_sum += loss.item() * bx.size(0)
            val_count += bx.size(0)

    val_loss = val_loss_sum / max(val_count, 1)

    print(f"[Epoch {epoch+1}] train_loss={train_loss:.8f}  val_loss={val_loss:.8f}")

    # ===== Early stopping =====
    if val_loss < best_val_loss - min_delta:
        best_val_loss = val_loss
        wait = 0
        torch.save(model.state_dict(), save_path)
        print(f"  -> Best model saved to {save_path}")
    else:
        wait += 1
        print(f"  -> no significant improvement ({wait}/{patience})")

        if wait >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

        
model.load_state_dict(torch.load(save_path, map_location=device))
model.eval()
print("Loaded best autoencoder.")

def extract_embeddings(model, loader, device):
    model.eval()
    embs = []

    with torch.no_grad():
        for (bx,) in tqdm(loader, desc="Extracting embeddings"):
            bx = bx.to(device, non_blocking=True)
            z = model.encode(bx)    # 只取 encoder 輸出
            embs.append(z.cpu())

    embs = torch.cat(embs).numpy()
    return embs

train_embeddings = extract_embeddings(model, train_loader, device)
test_embeddings  = extract_embeddings(model, test_loader, device)

print("train_embeddings shape:", train_embeddings.shape)
print("test_embeddings shape :", test_embeddings.shape)

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

emb_scaled = StandardScaler().fit_transform(train_embeddings)
emb_pca = PCA(n_components=20).fit_transform(emb_scaled)

print("PCA shape:", emb_pca.shape)
```

```python
#Autoencoder版本降維
import os
import gc
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm


import pandas as pd
from numpy.lib.stride_tricks import sliding_window_view
from sklearn.preprocessing import MinMaxScaler
import glob
import torch.nn.functional as F

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]


TIME_STEP = 126
time_step = 126
WINDOW_STRIDE = 5          # 建議先 5，降低重疊
TRAIN_RATIO = 0.8
BATCH_SIZE = 512
NUM_WORKERS = 0
LR = 5e-4
NUM_EPOCHS = 20
chunk_size = 50

PARQUET_GLOB = "parquet_data/*.parquet"
parquet_files = glob.glob(PARQUET_GLOB)

# ===== 版本 A 事件定義 =====
FUTURE_HIT_DAYS = 20       # 未來幾天內觀察是否 hit
HIT_PCT = 0.10             # hit 10%
USE_HIGH_FOR_HIT = False   # False=用 Close, True=用 High（更敏感，但較樂觀）

print(f"Found parquet files: {len(parquet_files)}")
print(f"Version A label: next {FUTURE_HIT_DAYS} days hit >= {HIT_PCT*100:.1f}%")
print(f"Use {'High' if USE_HIGH_FOR_HIT else 'Close'} for hit detection")


parquet_files = glob.glob("parquet_data/*.parquet")

# ========= Step 1: 全域 scaler =========
global_min, global_max = None, None
for pq in tqdm(parquet_files, desc="Global MinMax"):
    df = pd.read_parquet(pq)[FEATURE_COLS].astype(np.float32)
    
    # ⭐ NaN 修補（避免全 NaN）
    df = df.ffill().bfill().fillna(0)
    if df.empty:
        continue

    if global_min is None:
        global_min = df.min().to_numpy()
        global_max = df.max().to_numpy()
    else:
        global_min = np.minimum(global_min, df.min().to_numpy())
        global_max = np.maximum(global_max, df.max().to_numpy())

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

def build_hit_label_windows(
    feats_scaled,
    close_raw,
    high_raw=None,
    time_step=126,
    future_hit_days=20,
    hit_pct=0.50,
    window_stride=5,
    use_high_for_hit=False
):
    """
    單一股票版本：
    輸入:
      feats_scaled: (N, F)
      close_raw   : (N,)
      high_raw    : (N,) or None

    輸出:
      X_windows: (M, time_step, F)
      y_hit    : (M,) 0/1
    """
    n = len(feats_scaled)
    max_i = n - time_step - future_hit_days
    if max_i <= 0:
        return None, None

    # 先做所有有效 windows（view，不複製）
    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    windows = windows[:max_i]
    windows = windows[::window_stride]

    y_list = []

    for i in range(0, max_i, window_stride):
        # anchor = 輸入窗最後一天的 Close
        anchor_price = close_raw[i + time_step - 1]

        # 未來 future_hit_days 天內的價格路徑
        if use_high_for_hit and high_raw is not None:
            future_prices = high_raw[i + time_step : i + time_step + future_hit_days]
        else:
            future_prices = close_raw[i + time_step : i + time_step + future_hit_days]

        if len(future_prices) == 0:
            y_list.append(0)
            continue

        future_max = np.max(future_prices)
        future_return = (future_max / (anchor_price + 1e-12)) - 1.0

        y_list.append(1 if future_return >= hit_pct else 0)

    y_hit = np.asarray(y_list, dtype=np.int8)

    return windows.astype(np.float32, copy=False), y_hit.astype(np.float32, copy=False)


# ========= Step 2: Build windows for Version A (chunk-safe) 

def safe_read_parquet_features(pq, feature_cols, keep_date=False):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺欄位 -> None
    - 清理 NaN / inf
    """
    try:
        df = pd.read_parquet(pq)

        if df is None or df.empty:
            return None

        missing_cols = [c for c in feature_cols if c not in df.columns]
        if len(missing_cols) > 0:
            print(f"[SKIP missing cols] {os.path.basename(pq)} -> {missing_cols}")
            return None

        keep_cols = list(feature_cols)
        if keep_date and "Date" in df.columns:
            keep_cols.append("Date")

        df = df[keep_cols].copy()

        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty or len(df) == 0:
            return None

        return df

    except Exception as e:
        print(f"[SKIP read error] {os.path.basename(pq)} -> {e}")
        return None

X_chunks = []
Y_chunks = []

valid_files_step2 = 0
skipped_files_step2 = 0

# ===== 版本 A 事件定義 =====
FUTURE_HIT_DAYS = 20
HIT_PCT = 0.50
USE_HIGH_FOR_HIT = False   # False=用 Close，True=用 High

print("\n========= Step 2: Build windows for Version A (chunk-safe) =========")
print(f"TIME_STEP={TIME_STEP}, FUTURE_HIT_DAYS={FUTURE_HIT_DAYS}, WINDOW_STRIDE={WINDOW_STRIDE}, HIT_PCT={HIT_PCT}")

for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start + chunk_size)
    chunk_files = parquet_files[start:end]

    # 這一個 chunk 的暫存
    X_chunk_list = []
    Y_chunk_list = []

    for pq in chunk_files:
        df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

        if df is None:
            skipped_files_step2 += 1
            continue

        feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)
        close_raw = df["Close"].to_numpy(dtype=np.float32)

        if USE_HIGH_FOR_HIT and "High" in df.columns:
            high_raw = df["High"].to_numpy(dtype=np.float32)
        else:
            high_raw = None

        if len(feats) == 0 or len(close_raw) == 0:
            skipped_files_step2 += 1
            continue

        feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)

        feats_scaled = scaler.transform(feats)
        feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

        X_win, y_hit = build_hit_label_windows(
            feats_scaled=feats_scaled,
            close_raw=close_raw,
            high_raw=high_raw,
            time_step=TIME_STEP,
            future_hit_days=FUTURE_HIT_DAYS,
            hit_pct=HIT_PCT,
            window_stride=WINDOW_STRIDE,
            use_high_for_hit=USE_HIGH_FOR_HIT
        )

        if X_win is None or y_hit is None or len(X_win) == 0:
            skipped_files_step2 += 1
            continue

        X_chunk_list.append(X_win.astype(np.float32, copy=False))
        Y_chunk_list.append(y_hit.astype(np.float32, copy=False))
        valid_files_step2 += 1

        del df, feats, feats_scaled, close_raw, high_raw, X_win, y_hit
        gc.collect()

    # 這個 chunk 做完，先合併一次，降低峰值記憶體
    if len(X_chunk_list) > 0:
        X_chunk = np.concatenate(X_chunk_list, axis=0).astype(np.float32)
        Y_chunk = np.concatenate(Y_chunk_list, axis=0).astype(np.float32)

        X_chunks.append(X_chunk)
        Y_chunks.append(Y_chunk)

        print(f"[Chunk {start}:{end}] X_chunk shape={X_chunk.shape}, Y_chunk shape={Y_chunk.shape}")

        del X_chunk_list, Y_chunk_list, X_chunk, Y_chunk
        gc.collect()

if len(X_chunks) == 0 or len(Y_chunks) == 0:
    raise ValueError("Step 2 沒有產生任何有效樣本")

print("\n=== Step 2 summary ===")
print("valid_files_step2   :", valid_files_step2)
print("skipped_files_step2 :", skipped_files_step2)

# 最後合併全部 chunk
X_all = np.concatenate(X_chunks, axis=0).astype(np.float32)
Y_all = np.concatenate(Y_chunks, axis=0).astype(np.float32)

print("X_all shape:", X_all.shape)
print("Y_all shape:", Y_all.shape)
print("X_all NaN? ", np.isnan(X_all).any())
print("Y_all NaN? ", np.isnan(Y_all).any())
print("X_all min/max/std:", X_all.min(), X_all.max(), X_all.std())
print("Positive rate overall:", Y_all.mean())

# 釋放 chunk list
del X_chunks, Y_chunks
gc.collect()

cut = int(0.8 * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

assert X_train.ndim == 3 and X_train.shape[1] == 126 and X_train.shape[2] == 14
assert X_test.ndim == 3 and X_test.shape[1] == 126 and X_test.shape[2] == 14


batch_size = 512   # A4000 通常可先試 512；如果 OOM 就降到 256
num_workers = 0    # 你的資料很大，先用 0 比較穩

train_ds = TensorDataset(X_train)
test_ds  = TensorDataset(X_test)

train_loader = DataLoader(
    train_ds,
    batch_size=batch_size,
    shuffle=True,
    num_workers=num_workers,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=batch_size,
    shuffle=False,
    num_workers=num_workers,
    pin_memory=True
)

class ConvSeqAutoencoder(nn.Module):
    def __init__(self, input_dim=14, seq_len=126, latent_dim=64):
        super().__init__()
        self.input_dim = input_dim
        self.seq_len = seq_len
        self.latent_dim = latent_dim

        # -------- Encoder --------
        # 輸入 shape: (B, 14, 126)
        self.encoder = nn.Sequential(
            nn.Conv1d(input_dim, 32, kernel_size=5, stride=2, padding=2),   # 126 -> 63
            nn.ReLU(),
            nn.Conv1d(32, 64, kernel_size=5, stride=2, padding=2),          # 63 -> 32
            nn.ReLU(),
            nn.Conv1d(64, 128, kernel_size=5, stride=2, padding=2),         # 32 -> 16
            nn.ReLU(),
        )

        # 最後會是 (B, 128, 16)
        self.flatten_dim = 128 * 16
        self.to_latent = nn.Linear(self.flatten_dim, latent_dim)

        # -------- Decoder --------
        self.from_latent = nn.Linear(latent_dim, self.flatten_dim)

        self.decoder = nn.Sequential(
            # (B, 128, 16)
            nn.ConvTranspose1d(128, 64, kernel_size=4, stride=2, padding=1),           # 16 -> 32
            nn.ReLU(),
            nn.ConvTranspose1d(64, 32, kernel_size=5, stride=2, padding=2),            # 32 -> 63
            nn.ReLU(),
            nn.ConvTranspose1d(32, input_dim, kernel_size=4, stride=2, padding=1),     # 63 -> 126
        )

    def encode(self, x):
        """
        x: (B, 126, 14)
        return: z -> (B, latent_dim)
        """
        x = x.transpose(1, 2)          # -> (B, 14, 126)
        h = self.encoder(x)            # -> (B, 128, 16)
        h = h.reshape(h.size(0), -1)   # -> (B, 2048)
        z = self.to_latent(h)          # -> (B, latent_dim)
        return z

    def decode(self, z):
        """
        z: (B, latent_dim)
        return x_hat: (B, 126, 14)
        """
        h = self.from_latent(z)        # -> (B, 2048)
        h = h.view(h.size(0), 128, 16) # -> (B, 128, 16)
        x_hat = self.decoder(h)        # -> (B, 14, 126)
        x_hat = x_hat.transpose(1, 2)  # -> (B, 126, 14)
        return x_hat

    def forward(self, x):
        z = self.encode(x)
        x_hat = self.decode(z)
        return x_hat, z
    
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

model = ConvSeqAutoencoder(
    input_dim=14,
    seq_len=126,
    latent_dim=64   # 你可改成 32 / 64 / 128
).to(device)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)


num_epochs = 30
best_val_loss = float("inf")
patience = 3
wait = 0
min_delta = 1e-7   # 改善小於這個就不算有進步

save_path = "best_autoencoder.pth"

for epoch in range(num_epochs):
    # ===== Train =====
    model.train()
    train_loss_sum = 0.0
    train_count = 0

    pbar = tqdm(train_loader, desc=f"AE Epoch {epoch+1}/{num_epochs}")

    for (bx,) in pbar:
        bx = bx.to(device, non_blocking=True)

        x_hat, z = model(bx)
        loss = criterion(x_hat, bx)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss_sum += loss.item() * bx.size(0)
        train_count += bx.size(0)

        pbar.set_postfix(loss=float(loss))

    train_loss = train_loss_sum / max(train_count, 1)

    # ===== Validation =====
    model.eval()
    val_loss_sum = 0.0
    val_count = 0

    with torch.no_grad():
        for (bx,) in test_loader:
            bx = bx.to(device, non_blocking=True)
            x_hat, z = model(bx)
            loss = criterion(x_hat, bx)

            val_loss_sum += loss.item() * bx.size(0)
            val_count += bx.size(0)

    val_loss = val_loss_sum / max(val_count, 1)

    print(f"[Epoch {epoch+1}] train_loss={train_loss:.8f}  val_loss={val_loss:.8f}")

    # ===== Early stopping =====
    if val_loss < best_val_loss - min_delta:
        best_val_loss = val_loss
        wait = 0
        torch.save(model.state_dict(), save_path)
        print(f"  -> Best model saved to {save_path}")
    else:
        wait += 1
        print(f"  -> no significant improvement ({wait}/{patience})")

        if wait >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            break

        
model.load_state_dict(torch.load(save_path, map_location=device))
model.eval()
print("Loaded best autoencoder.")

def extract_embeddings(model, loader, device):
    model.eval()
    embs = []

    with torch.no_grad():
        for (bx,) in tqdm(loader, desc="Extracting embeddings"):
            bx = bx.to(device, non_blocking=True)
            z = model.encode(bx)    # 只取 encoder 輸出
            embs.append(z.cpu())

    embs = torch.cat(embs).numpy()
    return embs

train_embeddings = extract_embeddings(model, train_loader, device)
test_embeddings  = extract_embeddings(model, test_loader, device)

print("train_embeddings shape:", train_embeddings.shape)
print("test_embeddings shape :", test_embeddings.shape)

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

emb_scaled = StandardScaler().fit_transform(train_embeddings)
emb_pca = PCA(n_components=20).fit_transform(emb_scaled)

print("PCA shape:", emb_pca.shape)
```
