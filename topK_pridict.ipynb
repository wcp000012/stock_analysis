```python
#2026.5.7 原來是global_min, global_max變成NaN 才會造成training的結果都怪怪的
#contrastive encoder 結果不好
# 現在看起來用regression然後找出topK的方式應該比較好
import os
import gc
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

TIME_STEP = 126
FUTURE_DAYS = 60          # ✅ 你指定要 60 天
WINDOW_STRIDE = 5         # 建議先 5，降低重疊與記憶體
TRAIN_RATIO = 0.8
BATCH_SIZE = 512
NUM_WORKERS = 0
LR = 5e-4
NUM_EPOCHS = 30

PARQUET_GLOB = "parquet_data/*.parquet"
parquet_files = glob.glob(PARQUET_GLOB)

USE_HIGH_FOR_MAX = False   # False=用 Close 的 future max，True=用 High 的 future max
print(f"Found parquet files: {len(parquet_files)}")
print(f"Future max log return horizon = {FUTURE_DAYS} days")
print(f"Use {'High' if USE_HIGH_FOR_MAX else 'Close'} for future max")
def safe_read_parquet_features(pq, feature_cols, keep_date=False):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺欄位 -> None
    - 清理 NaN / inf
    """
    try:
        df = pd.read_parquet(pq)

        if df is None or df.empty:
            return None

        missing_cols = [c for c in feature_cols if c not in df.columns]
        if len(missing_cols) > 0:
            print(f"[SKIP missing cols] {os.path.basename(pq)} -> {missing_cols}")
            return None

        keep_cols = list(feature_cols)
        if keep_date and "Date" in df.columns:
            keep_cols.append("Date")

        df = df[keep_cols].copy()

        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty or len(df) == 0:
            return None

        return df

    except Exception as e:
        print(f"[SKIP read error] {os.path.basename(pq)} -> {e}")
        return None
global_min, global_max = None, None
valid_files_step1 = 0
skipped_files_step1 = 0

for pq in tqdm(parquet_files, desc="Step1 Global MinMax"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step1 += 1
        continue

    arr = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

    mn = np.nanmin(arr, axis=0)
    mx = np.nanmax(arr, axis=0)

    if not np.isfinite(mn).all() or not np.isfinite(mx).all():
        skipped_files_step1 += 1
        continue

    if global_min is None:
        global_min = mn
        global_max = mx
    else:
        global_min = np.minimum(global_min, mn)
        global_max = np.maximum(global_max, mx)

    valid_files_step1 += 1

if valid_files_step1 == 0:
    raise ValueError("Step 1 沒有任何有效 parquet 可建立 scaler")

print("\n=== Step 1 summary ===")
print("valid_files_step1   :", valid_files_step1)
print("skipped_files_step1 :", skipped_files_step1)

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

assert np.isfinite(scaler.data_min_).all()
assert np.isfinite(scaler.data_max_).all()
assert np.isfinite(scaler.data_range_).all()
def build_future_max_log_return_windows(
    feats_scaled,
    close_raw,
    high_raw=None,
    time_step=126,
    future_days=60,
    window_stride=5,
    use_high_for_max=False
):
    """
    單一股票版本：
    輸入:
      feats_scaled: (N, F)
      close_raw   : (N,)
      high_raw    : (N,) or None

    輸出:
      X_windows: (M, time_step, F)
      y_max_lr : (M,)  連續值，未來 future_days 內的最大 log return
    """
    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0:
        return None, None

    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    windows = windows[:max_i]
    windows = windows[::window_stride]

    y_list = []

    for i in range(0, max_i, window_stride):
        # anchor = 輸入窗最後一天的 Close
        anchor_price = close_raw[i + time_step - 1]

        # 未來觀察區間
        if use_high_for_max and high_raw is not None:
            future_prices = high_raw[i + time_step : i + time_step + future_days]
        else:
            future_prices = close_raw[i + time_step : i + time_step + future_days]

        if len(future_prices) == 0:
            y_list.append(0.0)
            continue

        future_max = np.max(future_prices)
        future_max_log_return = np.log((future_max + 1e-12) / (anchor_price + 1e-12))

        y_list.append(future_max_log_return)

    y_max_lr = np.asarray(y_list, dtype=np.float32)
    return windows.astype(np.float32, copy=False), y_max_lr
X_chunks = []
Y_chunks = []

valid_files_step2 = 0
skipped_files_step2 = 0
chunk_size = 20   # 建議先小一點，記憶體比較穩

print("\n========= Step 2: Build windows for future max log return regression =========")
print(f"TIME_STEP={TIME_STEP}, FUTURE_DAYS={FUTURE_DAYS}, WINDOW_STRIDE={WINDOW_STRIDE}")

for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start + chunk_size)
    chunk_files = parquet_files[start:end]

    X_chunk_list = []
    Y_chunk_list = []

    for pq in chunk_files:
        df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

        if df is None:
            skipped_files_step2 += 1
            continue

        feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)
        close_raw = df["Close"].to_numpy(dtype=np.float32)

        if USE_HIGH_FOR_MAX and "High" in df.columns:
            high_raw = df["High"].to_numpy(dtype=np.float32)
        else:
            high_raw = None

        if len(feats) == 0 or len(close_raw) == 0:
            skipped_files_step2 += 1
            continue

        feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)

        feats_scaled = scaler.transform(feats)
        feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

        X_win, y_max_lr = build_future_max_log_return_windows(
            feats_scaled=feats_scaled,
            close_raw=close_raw,
            high_raw=high_raw,
            time_step=TIME_STEP,
            future_days=FUTURE_DAYS,
            window_stride=WINDOW_STRIDE,
            use_high_for_max=USE_HIGH_FOR_MAX
        )

        if X_win is None or y_max_lr is None or len(X_win) == 0:
            skipped_files_step2 += 1
            continue

        X_chunk_list.append(X_win.astype(np.float32, copy=False))
        Y_chunk_list.append(y_max_lr.astype(np.float32, copy=False))
        valid_files_step2 += 1

        del df, feats, feats_scaled, close_raw, high_raw, X_win, y_max_lr
        gc.collect()

    if len(X_chunk_list) > 0:
        X_chunk = np.concatenate(X_chunk_list, axis=0).astype(np.float32)
        Y_chunk = np.concatenate(Y_chunk_list, axis=0).astype(np.float32)

        X_chunks.append(X_chunk)
        Y_chunks.append(Y_chunk)

        print(f"[Chunk {start}:{end}] X_chunk shape={X_chunk.shape}, Y_chunk shape={Y_chunk.shape}")

        del X_chunk_list, Y_chunk_list, X_chunk, Y_chunk
        gc.collect()

if len(X_chunks) == 0 or len(Y_chunks) == 0:
    raise ValueError("Step 2 沒有產生任何有效樣本")

print("\n=== Step 2 summary ===")
print("valid_files_step2   :", valid_files_step2)
print("skipped_files_step2 :", skipped_files_step2)

X_all = np.concatenate(X_chunks, axis=0).astype(np.float32)
Y_all = np.concatenate(Y_chunks, axis=0).astype(np.float32)

print("X_all shape:", X_all.shape)
print("Y_all shape:", Y_all.shape)
print("X_all NaN? ", np.isnan(X_all).any())
print("Y_all NaN? ", np.isnan(Y_all).any())
print("X_all min/max/std:", X_all.min(), X_all.max(), X_all.std())
print("Y_all min/max/std:", Y_all.min(), Y_all.max(), Y_all.std())
print("Y_all quantiles:", np.quantile(Y_all, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))

del X_chunks, Y_chunks
gc.collect()
cut = int(TRAIN_RATIO * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()

print("\n=== Train / Test ===")
print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
print("X_test shape :", X_test.shape)
print("Y_test shape :", Y_test.shape)

train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test, Y_test)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)
class PatchTSTRegression(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,
        stride=16,
        d_model=128,
        n_heads=4,
        n_layers=3,
        dropout=0.1
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        patch_dim = patch_len * input_size
        self.patch_embed = nn.Linear(patch_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        B, T, F = x.shape

        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)
        h = self.patch_embed(patches)
        z = self.transformer(h)

        z_last = z[:, -1, :]
        out = self.head(z_last).squeeze(-1)
        return out
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

model = PatchTSTRegression(input_size=len(FEATURE_COLS)).to(device)

criterion = nn.SmoothL1Loss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
for epoch in range(NUM_EPOCHS):
    model.train()
    train_loss_sum = 0.0
    train_count = 0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")

    for bx, by in pbar:
        bx = bx.to(device, non_blocking=True)
        by = by.to(device, non_blocking=True)

        pred = model(bx)
        loss = criterion(pred, by)

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        train_loss_sum += loss.item() * bx.size(0)
        train_count += bx.size(0)

        pbar.set_postfix(batch_loss=float(loss))

    train_loss = train_loss_sum / max(train_count, 1)

    # validation
    model.eval()
    val_loss_sum = 0.0
    val_count = 0

    with torch.no_grad():
        for bx, by in test_loader:
            bx = bx.to(device, non_blocking=True)
            by = by.to(device, non_blocking=True)

            pred = model(bx)
            loss = criterion(pred, by)

            val_loss_sum += loss.item() * bx.size(0)
            val_count += bx.size(0)

    val_loss = val_loss_sum / max(val_count, 1)

    print(f"[Epoch {epoch+1}] train_loss={train_loss:.6f}  val_loss={val_loss:.6f}")

    model.eval()
preds = []
trues = []

with torch.no_grad():
    for bx, by in test_loader:
        bx = bx.to(device, non_blocking=True)
        pred_batch = model(bx).cpu()
        preds.append(pred_batch)
        trues.append(by)

pred = torch.cat(preds).numpy()
true = torch.cat(trues).numpy()

print("\n=== Eval sanity check ===")
print("pred NaN?", np.isnan(pred).any())
print("true NaN?", np.isnan(true).any())
print("pred std:", pred.std())
print("true std:", true.std())
print("pred quantiles:", np.quantile(pred, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))
print("true quantiles:", np.quantile(true, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))

def topk_regression_evaluation(
    pred,
    true,
    k_list=(0.01, 0.02, 0.05, 0.10),
    hit_thresholds=(0.10, 0.15, 0.20)
):
    """
    pred: 模型預測的 future max log return
    true: 真實的 future max log return

    hit_thresholds: 以 simple return 門檻來做 hit rate@K
    """
    pred = np.asarray(pred).reshape(-1)
    true = np.asarray(true).reshape(-1)

    n = len(pred)
    order = np.argsort(-pred)

    # 真實的 simple return（從 log return 轉回）
    true_simple = np.expm1(true)

    rows = []

    for k in k_list:
        top_n = max(1, int(np.floor(n * k)))
        top_idx = order[:top_n]

        row = {
            "top_k_ratio": k,
            "top_n": top_n,
            "avg_pred_log_return": float(pred[top_idx].mean()),
            "avg_true_log_return": float(true[top_idx].mean()),
            "median_true_log_return": float(np.median(true[top_idx])),
            "avg_true_simple_return": float(true_simple[top_idx].mean()),
            "median_true_simple_return": float(np.median(true_simple[top_idx])),
        }

        for thr in hit_thresholds:
            hit_rate = (true_simple[top_idx] >= thr).mean()
            base_rate = (true_simple >= thr).mean()
            lift = hit_rate / (base_rate + 1e-12)

            row[f"hit_rate_ge_{int(thr*100)}pct"] = float(hit_rate)
            row[f"base_rate_ge_{int(thr*100)}pct"] = float(base_rate)
            row[f"lift_ge_{int(thr*100)}pct"] = float(lift)

        rows.append(row)

    return pd.DataFrame(rows)
def decile_analysis_regression(pred, true, n_bins=10):
    pred = np.asarray(pred).reshape(-1)
    true = np.asarray(true).reshape(-1)

    true_simple = np.expm1(true)

    order = np.argsort(-pred)
    bins = np.array_split(order, n_bins)

    rows = []
    for i, idx in enumerate(bins, start=1):
        rows.append({
            "bin_rank": i,   # 1 = highest score bin
            "count": len(idx),
            "avg_pred_log_return": float(pred[idx].mean()),
            "avg_true_log_return": float(true[idx].mean()),
            "avg_true_simple_return": float(true_simple[idx].mean()),
            "median_true_simple_return": float(np.median(true_simple[idx])),
            "hit_rate_ge_10pct": float((true_simple[idx] >= 0.10).mean()),
            "hit_rate_ge_15pct": float((true_simple[idx] >= 0.15).mean()),
            "hit_rate_ge_20pct": float((true_simple[idx] >= 0.20).mean()),
        })

    return pd.DataFrame(rows)
# ===== Top-K table =====
topk_df = topk_regression_evaluation(
    pred=pred,
    true=true,
    k_list=(0.01, 0.02, 0.05, 0.10),
    hit_thresholds=(0.10, 0.15, 0.20)
)

print("\n=== Top-K Regression Evaluation ===")
print(topk_df.to_string(index=False))

# ===== Decile table =====
decile_df = decile_analysis_regression(pred, true, n_bins=10)

print("\n=== Decile Analysis (1 = highest score bin) ===")
print(decile_df.to_string(index=False))

# ===== Plot 1: Top-K future return =====
plt.figure(figsize=(8, 5))
plt.plot(topk_df["top_k_ratio"] * 100, topk_df["avg_true_simple_return"], marker="o", label="Avg future max return")
plt.plot(topk_df["top_k_ratio"] * 100, topk_df["median_true_simple_return"], marker="o", label="Median future max return")
plt.axhline(np.expm1(true).mean(), color="gray", linestyle="--", label="All-sample avg")
plt.xlabel("Top-K (%)")
plt.ylabel("Future max simple return")
plt.title("Top-K Future Max Return")
plt.legend()
plt.grid(True)
plt.show()

# ===== Plot 2: Decile spread =====
plt.figure(figsize=(8, 5))
plt.plot(decile_df["bin_rank"], decile_df["avg_true_simple_return"], marker="o", label="Avg future max return")
plt.plot(decile_df["bin_rank"], decile_df["hit_rate_ge_10pct"], marker="o", label="Hit rate >=10%")
plt.plot(decile_df["bin_rank"], decile_df["hit_rate_ge_15pct"], marker="o", label="Hit rate >=15%")
plt.xlabel("Score bin rank (1 = highest)")
plt.ylabel("Metric")
plt.title("Decile Spread by Regression Score")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#這版已經十分接近實戰的架構了，但是因為資料是以所有時間中隨機的80:20來切分，而非以一段時間往後做預測，成效未必可信
import os
import gc
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

TIME_STEP = 126
FUTURE_DAYS = 60
WINDOW_STRIDE = 5
TRAIN_RATIO = 0.8
BATCH_SIZE = 512
NUM_WORKERS = 0
LR = 5e-4
NUM_EPOCHS = 30

PARQUET_GLOB = "parquet_data/*.parquet"
parquet_files = glob.glob(PARQUET_GLOB)

USE_HIGH_FOR_MAX = False

print(f"Found parquet files: {len(parquet_files)}")
def safe_read_parquet_features(pq, feature_cols, keep_date=False):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺欄位 -> None
    - 清理 NaN / inf
    - 若 keep_date=True 且有 Date，則保留並排序
    """
    try:
        df = pd.read_parquet(pq)

        if df is None or df.empty:
            return None

        missing_cols = [c for c in feature_cols if c not in df.columns]
        if len(missing_cols) > 0:
            print(f"[SKIP missing cols] {os.path.basename(pq)} -> {missing_cols}")
            return None

        keep_cols = list(feature_cols)
        if keep_date and "Date" in df.columns:
            keep_cols.append("Date")

        df = df[keep_cols].copy()

        # Date 處理
        if keep_date and "Date" in df.columns:
            df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
            df = df.sort_values("Date").reset_index(drop=True)

        # 數值化
        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty or len(df) == 0:
            return None

        return df

    except Exception as e:
        print(f"[SKIP read error] {os.path.basename(pq)} -> {e}")
        return None
global_min, global_max = None, None
valid_files_step1 = 0
skipped_files_step1 = 0

for pq in tqdm(parquet_files, desc="Step1 Global MinMax"):
    df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=False)

    if df is None:
        skipped_files_step1 += 1
        continue

    arr = df[FEATURE_COLS].to_numpy(dtype=np.float32)
    arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

    mn = np.nanmin(arr, axis=0)
    mx = np.nanmax(arr, axis=0)

    if not np.isfinite(mn).all() or not np.isfinite(mx).all():
        skipped_files_step1 += 1
        continue

    if global_min is None:
        global_min = mn
        global_max = mx
    else:
        global_min = np.minimum(global_min, mn)
        global_max = np.maximum(global_max, mx)

    valid_files_step1 += 1

if valid_files_step1 == 0:
    raise ValueError("Step 1 沒有任何有效 parquet 可建立 scaler")

print("\n=== Step 1 summary ===")
print("valid_files_step1   :", valid_files_step1)
print("skipped_files_step1 :", skipped_files_step1)

scaler = MinMaxScaler()
scaler.fit(np.vstack([global_min, global_max]))

assert np.isfinite(scaler.data_min_).all()
assert np.isfinite(scaler.data_max_).all()
assert np.isfinite(scaler.data_range_).all()
def build_future_max_log_return_windows_with_meta(
    feats_scaled,
    close_raw,
    high_raw=None,
    date_raw=None,
    stock_code=None,
    time_step=126,
    future_days=60,
    window_stride=5,
    use_high_for_max=False
):
    """
    單一股票版本，回傳：
      X_windows: (M, time_step, F)
      y_max_lr : (M,)
      meta_df  : DataFrame with stock_code / anchor_date / anchor_idx
    """
    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0:
        return None, None, None

    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    windows = windows[:max_i]
    windows = windows[::window_stride]

    y_list = []
    meta_rows = []

    for i in range(0, max_i, window_stride):
        anchor_idx = i + time_step - 1
        anchor_price = close_raw[anchor_idx]

        if use_high_for_max and high_raw is not None:
            future_prices = high_raw[i + time_step : i + time_step + future_days]
        else:
            future_prices = close_raw[i + time_step : i + time_step + future_days]

        if len(future_prices) == 0:
            y_list.append(0.0)
        else:
            future_max = np.max(future_prices)
            future_max_log_return = np.log((future_max + 1e-12) / (anchor_price + 1e-12))
            y_list.append(future_max_log_return)

        anchor_date = pd.NaT
        if date_raw is not None and anchor_idx < len(date_raw):
            anchor_date = date_raw[anchor_idx]

        meta_rows.append({
            "stock_code": stock_code,
            "anchor_idx": anchor_idx,
            "anchor_date": anchor_date
        })

    y_max_lr = np.asarray(y_list, dtype=np.float32)
    meta_df = pd.DataFrame(meta_rows)

    return windows.astype(np.float32, copy=False), y_max_lr, meta_df
X_chunks = []
Y_chunks = []
META_chunks = []

valid_files_step2 = 0
skipped_files_step2 = 0
chunk_size = 20   # 可依 RAM 調整

print("\n========= Step 2: Build windows + metadata =========")
print(f"TIME_STEP={TIME_STEP}, FUTURE_DAYS={FUTURE_DAYS}, WINDOW_STRIDE={WINDOW_STRIDE}")

for start in range(0, len(parquet_files), chunk_size):
    end = min(len(parquet_files), start + chunk_size)
    chunk_files = parquet_files[start:end]

    X_chunk_list = []
    Y_chunk_list = []
    META_chunk_list = []

    for pq in chunk_files:
        df = safe_read_parquet_features(pq, FEATURE_COLS, keep_date=True)

        if df is None:
            skipped_files_step2 += 1
            continue

        stock_code = os.path.basename(pq).replace(".parquet", "")

        feats = df[FEATURE_COLS].to_numpy(dtype=np.float32)
        close_raw = df["Close"].to_numpy(dtype=np.float32)

        if USE_HIGH_FOR_MAX and "High" in df.columns:
            high_raw = df["High"].to_numpy(dtype=np.float32)
        else:
            high_raw = None

        date_raw = df["Date"].to_numpy() if "Date" in df.columns else None

        if len(feats) == 0 or len(close_raw) == 0:
            skipped_files_step2 += 1
            continue

        feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)

        feats_scaled = scaler.transform(feats)
        feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

        X_win, y_max_lr, meta_df = build_future_max_log_return_windows_with_meta(
            feats_scaled=feats_scaled,
            close_raw=close_raw,
            high_raw=high_raw,
            date_raw=date_raw,
            stock_code=stock_code,
            time_step=TIME_STEP,
            future_days=FUTURE_DAYS,
            window_stride=WINDOW_STRIDE,
            use_high_for_max=USE_HIGH_FOR_MAX
        )

        if X_win is None or y_max_lr is None or len(X_win) == 0:
            skipped_files_step2 += 1
            continue

        X_chunk_list.append(X_win.astype(np.float32, copy=False))
        Y_chunk_list.append(y_max_lr.astype(np.float32, copy=False))
        META_chunk_list.append(meta_df)

        valid_files_step2 += 1

        del df, feats, feats_scaled, close_raw, high_raw, date_raw, X_win, y_max_lr, meta_df
        gc.collect()

    if len(X_chunk_list) > 0:
        X_chunk = np.concatenate(X_chunk_list, axis=0).astype(np.float32)
        Y_chunk = np.concatenate(Y_chunk_list, axis=0).astype(np.float32)
        META_chunk = pd.concat(META_chunk_list, ignore_index=True)

        X_chunks.append(X_chunk)
        Y_chunks.append(Y_chunk)
        META_chunks.append(META_chunk)

        print(f"[Chunk {start}:{end}] X_chunk={X_chunk.shape}, Y_chunk={Y_chunk.shape}, META_chunk={META_chunk.shape}")

        del X_chunk_list, Y_chunk_list, META_chunk_list, X_chunk, Y_chunk, META_chunk
        gc.collect()

if len(X_chunks) == 0 or len(Y_chunks) == 0:
    raise ValueError("Step 2 沒有產生任何有效樣本")

print("\n=== Step 2 summary ===")
print("valid_files_step2   :", valid_files_step2)
print("skipped_files_step2 :", skipped_files_step2)

X_all = np.concatenate(X_chunks, axis=0).astype(np.float32)
Y_all = np.concatenate(Y_chunks, axis=0).astype(np.float32)
META_all = pd.concat(META_chunks, ignore_index=True)

print("X_all shape:", X_all.shape)
print("Y_all shape:", Y_all.shape)
print("META_all shape:", META_all.shape)
print("X_all NaN? ", np.isnan(X_all).any())
print("Y_all NaN? ", np.isnan(Y_all).any())
print("X_all min/max/std:", X_all.min(), X_all.max(), X_all.std())
print("Y_all min/max/std:", Y_all.min(), Y_all.max(), Y_all.std())
print("Y_all quantiles:", np.quantile(Y_all, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))

assert len(X_all) == len(Y_all) == len(META_all), "X/Y/META 長度不一致"

del X_chunks, Y_chunks, META_chunks
gc.collect()
cut = int(TRAIN_RATIO * len(Y_all))

X_train = torch.from_numpy(X_all[:cut]).contiguous()
Y_train = torch.from_numpy(Y_all[:cut]).contiguous()
META_train = META_all.iloc[:cut].reset_index(drop=True)

X_test  = torch.from_numpy(X_all[cut:]).contiguous()
Y_test  = torch.from_numpy(Y_all[cut:]).contiguous()
META_test = META_all.iloc[cut:].reset_index(drop=True)

print("\n=== Train / Test ===")
print("X_train shape:", X_train.shape)
print("Y_train shape:", Y_train.shape)
print("META_train shape:", META_train.shape)
print("X_test shape :", X_test.shape)
print("Y_test shape :", Y_test.shape)
print("META_test shape:", META_test.shape)

assert len(X_train) == len(Y_train) == len(META_train)
assert len(X_test) == len(Y_test) == len(META_test)

train_ds = TensorDataset(X_train, Y_train)
test_ds  = TensorDataset(X_test, Y_test)

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)
class PatchTSTRegression(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,
        stride=16,
        d_model=128,
        n_heads=4,
        n_layers=3,
        dropout=0.1
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        patch_dim = patch_len * input_size
        self.patch_embed = nn.Linear(patch_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        B, T, F = x.shape

        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)
        h = self.patch_embed(patches)
        z = self.transformer(h)

        z_last = z[:, -1, :]
        out = self.head(z_last).squeeze(-1)
        return out
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

model = PatchTSTRegression(input_size=len(FEATURE_COLS)).to(device)

criterion = nn.SmoothL1Loss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
for epoch in range(NUM_EPOCHS):
    model.train()
    train_loss_sum = 0.0
    train_count = 0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{NUM_EPOCHS}")

    for bx, by in pbar:
        bx = bx.to(device, non_blocking=True)
        by = by.to(device, non_blocking=True)

        pred = model(bx)
        loss = criterion(pred, by)

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        train_loss_sum += loss.item() * bx.size(0)
        train_count += bx.size(0)

        pbar.set_postfix(batch_loss=float(loss))

    train_loss = train_loss_sum / max(train_count, 1)

    # validation
    model.eval()
    val_loss_sum = 0.0
    val_count = 0

    with torch.no_grad():
        for bx, by in test_loader:
            bx = bx.to(device, non_blocking=True)
            by = by.to(device, non_blocking=True)

            pred = model(bx)
            loss = criterion(pred, by)

            val_loss_sum += loss.item() * bx.size(0)
            val_count += bx.size(0)

    val_loss = val_loss_sum / max(val_count, 1)

    print(f"[Epoch {epoch+1}] train_loss={train_loss:.6f}  val_loss={val_loss:.6f}")

    model.eval()
preds = []
trues = []

with torch.no_grad():
    for bx, by in test_loader:
        bx = bx.to(device, non_blocking=True)
        pred_batch = model(bx).cpu()
        preds.append(pred_batch)
        trues.append(by)

pred = torch.cat(preds).numpy()
true = torch.cat(trues).numpy()

print("\n=== Eval sanity check ===")
print("pred NaN?", np.isnan(pred).any())
print("true NaN?", np.isnan(true).any())
print("pred std:", pred.std())
print("true std:", true.std())
print("pred quantiles:", np.quantile(pred, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))
print("true quantiles:", np.quantile(true, [0, 0.5, 0.9, 0.95, 0.99, 1.0]))

pred_df = META_test.copy()

pred_df["pred_log_return"] = pred
pred_df["true_log_return"] = true

pred_df["pred_simple_return"] = np.expm1(pred_df["pred_log_return"])
pred_df["true_simple_return"] = np.expm1(pred_df["true_log_return"])

# 方便後續看 hit rate
pred_df["hit_ge_10pct"] = (pred_df["true_simple_return"] >= 0.10).astype(int)
pred_df["hit_ge_15pct"] = (pred_df["true_simple_return"] >= 0.15).astype(int)
pred_df["hit_ge_20pct"] = (pred_df["true_simple_return"] >= 0.20).astype(int)

print(pred_df.head())
print(pred_df.shape)
def build_daily_watchlist(pred_df, top_n=10):
    """
    依 anchor_date 分組，每天選 pred_log_return 最高的前 top_n 筆
    """
    df = pred_df.copy()

    # 如果 anchor_date 有 NaT，先丟掉
    df = df.dropna(subset=["anchor_date"]).copy()
    df["anchor_date"] = pd.to_datetime(df["anchor_date"])

    # 保險：同一天同一股票若有重複，取最高分
    df = (
        df.groupby(["anchor_date", "stock_code"], as_index=False)
          .agg({
              "pred_log_return": "max",
              "pred_simple_return": "max",
              "true_log_return": "max",
              "true_simple_return": "max",
              "hit_ge_10pct": "max",
              "hit_ge_15pct": "max",
              "hit_ge_20pct": "max",
          })
    )

    # 每天排序後取前 top_n
    daily_top = (
        df.sort_values(["anchor_date", "pred_log_return"], ascending=[True, False])
          .groupby("anchor_date", group_keys=False)
          .head(top_n)
          .reset_index(drop=True)
    )

    return daily_top

daily_watchlist_top10 = build_daily_watchlist(pred_df, top_n=10)
daily_watchlist_top20 = build_daily_watchlist(pred_df, top_n=20)

print("\n=== Daily watchlist Top 10 ===")
print(daily_watchlist_top10.head(20))
def build_weekly_watchlist(pred_df, top_n=10):
    df = pred_df.copy()

    df = df.dropna(subset=["anchor_date"]).copy()
    df["anchor_date"] = pd.to_datetime(df["anchor_date"])
    df["week"] = df["anchor_date"].dt.to_period("W").astype(str)

    # 同一週同一股票若有重複，取最高分
    df = (
        df.groupby(["week", "stock_code"], as_index=False)
          .agg({
              "anchor_date": "max",
              "pred_log_return": "max",
              "pred_simple_return": "max",
              "true_log_return": "max",
              "true_simple_return": "max",
              "hit_ge_10pct": "max",
              "hit_ge_15pct": "max",
              "hit_ge_20pct": "max",
          })
    )

    weekly_top = (
        df.sort_values(["week", "pred_log_return"], ascending=[True, False])
          .groupby("week", group_keys=False)
          .head(top_n)
          .reset_index(drop=True)
    )

    return weekly_top
weekly_watchlist_top10 = build_weekly_watchlist(pred_df, top_n=10)

print("\n=== Weekly watchlist Top")

def summarize_watchlist(watchlist_df, name="watchlist"):
    df = watchlist_df.copy()

    summary = {
        "name": name,
        "n_rows": len(df),
        "n_unique_stocks": df["stock_code"].nunique() if "stock_code" in df.columns else np.nan,
        "avg_true_log_return": float(df["true_log_return"].mean()),
        "median_true_log_return": float(df["true_log_return"].median()),
        "avg_true_simple_return": float(df["true_simple_return"].mean()),
        "median_true_simple_return": float(df["true_simple_return"].median()),
        "hit_rate_ge_10pct": float(df["hit_ge_10pct"].mean()),
        "hit_rate_ge_15pct": float(df["hit_ge_15pct"].mean()),
        "hit_rate_ge_20pct": float(df["hit_ge_20pct"].mean()),
        "avg_pred_log_return": float(df["pred_log_return"].mean()),
    }
    return pd.Series(summary)

summary_daily_10 = summarize_watchlist(daily_watchlist_top10, "daily_top10")
summary_daily_20 = summarize_watchlist(daily_watchlist_top20, "daily_top20")
summary_weekly_10 = summarize_watchlist(weekly_watchlist_top10, "weekly_top10")

watchlist_summary = pd.DataFrame([
    summary_daily_10,
    summary_daily_20,
    summary_weekly_10
])

print("\n=== Watchlist Summary ===")
print(watchlist_summary.to_string(index=False))

def summarize_by_date(daily_watchlist_df):
    df = daily_watchlist_df.copy()

    by_date = (
        df.groupby("anchor_date")
          .agg(
              n=("stock_code", "size"),
              avg_true_simple_return=("true_simple_return", "mean"),
              median_true_simple_return=("true_simple_return", "median"),
              hit_ge_10pct=("hit_ge_10pct", "mean"),
              hit_ge_15pct=("hit_ge_15pct", "mean"),
              hit_ge_20pct=("hit_ge_20pct", "mean"),
              avg_pred_log_return=("pred_log_return", "mean"),
          )
          .reset_index()
          .sort_values("anchor_date")
    )
    return by_date
daily_summary_top10 = summarize_by_date(daily_watchlist_top10)

print("\n=== Daily Summary Top 10 ===")
print(daily_summary_top10.head())

plt.figure(figsize=(10,5))
plt.plot(daily_summary_top10["anchor_date"], daily_summary_top10["avg_true_simple_return"], label="Avg true simple return")
plt.plot(daily_summary_top10["anchor_date"], daily_summary_top10["hit_ge_10pct"], label="Hit rate >=10%")
plt.plot(daily_summary_top10["anchor_date"], daily_summary_top10["hit_ge_15pct"], label="Hit rate >=15%")
plt.xlabel("Date")
plt.ylabel("Metric")
plt.title("Daily Watchlist Performance (Top 10)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import os
import gc
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm
from numpy.lib.stride_tricks import sliding_window_view

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

FEATURE_COLS = [
    "Close","High","Low","Open","Volume","MA10","RSI",
    "margin_buy","margin_sell","short_buy","short_sell",
    "foreign_net","invest_net","dealer_net"
]

PARQUET_GLOB = "parquet_data/*.parquet"
parquet_files = glob.glob(PARQUET_GLOB)

TIME_STEP = 126
FUTURE_DAYS = 60
WINDOW_STRIDE = 5

BATCH_SIZE = 512
NUM_WORKERS = 0
LR = 5e-4
NUM_EPOCHS = 10   # walk-forward 先保守，避免每 fold 太久
USE_HIGH_FOR_MAX = False
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Found parquet files:", len(parquet_files))
print("Device:", DEVICE)

def safe_read_parquet_features(pq, feature_cols, require_date=True):
    """
    安全讀 parquet：
    - 空檔 -> None
    - 缺必要欄位 -> None
    - Date 轉 datetime 並排序
    - 清理 NaN / inf
    """
    try:
        df = pd.read_parquet(pq)

        if df is None or df.empty:
            return None

        needed = feature_cols.copy()
        if require_date:
            if "Date" not in df.columns:
                return None
            needed = needed + ["Date"]

        missing = [c for c in feature_cols if c not in df.columns]
        if len(missing) > 0:
            return None

        df = df[needed].copy()

        if require_date:
            df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
            df = df.sort_values("Date").reset_index(drop=True)

        for c in feature_cols:
            df[c] = pd.to_numeric(df[c], errors="coerce")

        df[feature_cols] = df[feature_cols].astype(np.float32)
        df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
        df[feature_cols] = df[feature_cols].ffill().bfill().fillna(0)

        if df.empty:
            return None

        return df

    except Exception:
        return None

def fit_global_scaler_until_date(parquet_files, feature_cols, train_end_date):
    """
    只用 Date <= train_end_date 的資料來建立 global scaler
    """
    global_min, global_max = None, None
    used_files = 0

    train_end_date = pd.Timestamp(train_end_date)

    for pq in tqdm(parquet_files, desc=f"Fit scaler until {train_end_date.date()}"):
        df = safe_read_parquet_features(pq, feature_cols, require_date=True)
        if df is None:
            continue

        df = df[df["Date"] <= train_end_date].copy()
        if df.empty:
            continue

        arr = df[feature_cols].to_numpy(dtype=np.float32)
        arr = np.nan_to_num(arr, nan=0.0, posinf=0.0, neginf=0.0)

        mn = np.nanmin(arr, axis=0)
        mx = np.nanmax(arr, axis=0)

        if not np.isfinite(mn).all() or not np.isfinite(mx).all():
            continue

        if global_min is None:
            global_min = mn
            global_max = mx
        else:
            global_min = np.minimum(global_min, mn)
            global_max = np.maximum(global_max, mx)

        used_files += 1

    if used_files == 0 or global_min is None:
        raise ValueError(f"無法在 {train_end_date.date()} 前建立 scaler")

    scaler = MinMaxScaler()
    scaler.fit(np.vstack([global_min, global_max]))

    return scaler
def build_future_max_log_return_windows_with_meta(
    feats_scaled,
    close_raw,
    high_raw=None,
    date_raw=None,
    stock_code=None,
    time_step=126,
    future_days=60,
    window_stride=5,
    use_high_for_max=False
):
    """
    回傳:
      X_windows: (M, time_step, F)
      y_max_lr : (M,)
      meta_df  : stock_code / anchor_idx / anchor_date / label_end_date
    """
    n = len(feats_scaled)
    max_i = n - time_step - future_days
    if max_i <= 0:
        return None, None, None

    windows = sliding_window_view(
        feats_scaled,
        (time_step, feats_scaled.shape[1])
    ).squeeze(1)

    windows = windows[:max_i]
    windows = windows[::window_stride]

    y_list = []
    meta_rows = []

    for i in range(0, max_i, window_stride):
        anchor_idx = i + time_step - 1
        label_end_idx = i + time_step + future_days - 1

        anchor_price = close_raw[anchor_idx]

        if use_high_for_max and high_raw is not None:
            future_prices = high_raw[i + time_step : i + time_step + future_days]
        else:
            future_prices = close_raw[i + time_step : i + time_step + future_days]

        if len(future_prices) == 0:
            y_val = 0.0
        else:
            future_max = np.max(future_prices)
            y_val = np.log((future_max + 1e-12) / (anchor_price + 1e-12))

        anchor_date = pd.NaT
        label_end_date = pd.NaT

        if date_raw is not None:
            if anchor_idx < len(date_raw):
                anchor_date = date_raw[anchor_idx]
            if label_end_idx < len(date_raw):
                label_end_date = date_raw[label_end_idx]

        y_list.append(y_val)
        meta_rows.append({
            "stock_code": stock_code,
            "anchor_idx": anchor_idx,
            "anchor_date": anchor_date,
            "label_end_date": label_end_date
        })

    y_max_lr = np.asarray(y_list, dtype=np.float32)
    meta_df = pd.DataFrame(meta_rows)

    return windows.astype(np.float32, copy=False), y_max_lr, meta_df
def build_dataset_for_period(
    parquet_files,
    scaler,
    feature_cols,
    train_end_date=None,
    test_start_date=None,
    test_end_date=None,
    mode="train",   # "train" or "test"
    time_step=126,
    future_days=60,
    window_stride=5,
    use_high_for_max=False,
    chunk_size=20
):
    """
    mode="train":
      保留 label_end_date <= train_end_date 的樣本

    mode="test":
      保留 test_start_date <= anchor_date <= test_end_date 的樣本
    """

    X_chunks = []
    Y_chunks = []
    META_chunks = []

    valid_files = 0
    skipped_files = 0

    train_end_date = pd.Timestamp(train_end_date) if train_end_date is not None else None
    test_start_date = pd.Timestamp(test_start_date) if test_start_date is not None else None
    test_end_date = pd.Timestamp(test_end_date) if test_end_date is not None else None

    for start in range(0, len(parquet_files), chunk_size):
        end = min(len(parquet_files), start + chunk_size)
        chunk_files = parquet_files[start:end]

        X_chunk_list = []
        Y_chunk_list = []
        META_chunk_list = []

        for pq in chunk_files:
            df = safe_read_parquet_features(pq, feature_cols, require_date=True)
            if df is None:
                skipped_files += 1
                continue

            stock_code = os.path.basename(pq).replace(".parquet", "")

            feats = df[feature_cols].to_numpy(dtype=np.float32)
            close_raw = df["Close"].to_numpy(dtype=np.float32)
            high_raw = df["High"].to_numpy(dtype=np.float32) if (use_high_for_max and "High" in df.columns) else None
            date_raw = df["Date"].to_numpy()

            if len(feats) == 0:
                skipped_files += 1
                continue

            feats = np.nan_to_num(feats, nan=0.0, posinf=0.0, neginf=0.0)
            feats_scaled = scaler.transform(feats)
            feats_scaled = np.nan_to_num(feats_scaled, nan=0.0, posinf=0.0, neginf=0.0)

            X_win, y_val, meta_df = build_future_max_log_return_windows_with_meta(
                feats_scaled=feats_scaled,
                close_raw=close_raw,
                high_raw=high_raw,
                date_raw=date_raw,
                stock_code=stock_code,
                time_step=time_step,
                future_days=future_days,
                window_stride=window_stride,
                use_high_for_max=use_high_for_max
            )

            if X_win is None or len(X_win) == 0:
                skipped_files += 1
                continue

            # 嚴格時間過濾
            if mode == "train":
                mask = meta_df["label_end_date"] <= train_end_date
            elif mode == "test":
                mask = (meta_df["anchor_date"] >= test_start_date) & (meta_df["anchor_date"] <= test_end_date)
            else:
                raise ValueError("mode must be 'train' or 'test'")

            if mask.sum() == 0:
                skipped_files += 1
                continue

            X_win = X_win[mask.to_numpy()]
            y_val = y_val[mask.to_numpy()]
            meta_df = meta_df.loc[mask].reset_index(drop=True)

            if len(X_win) == 0:
                skipped_files += 1
                continue

            X_chunk_list.append(X_win.astype(np.float32, copy=False))
            Y_chunk_list.append(y_val.astype(np.float32, copy=False))
            META_chunk_list.append(meta_df)

            valid_files += 1

            del df, feats, feats_scaled, close_raw, high_raw, date_raw, X_win, y_val, meta_df
            gc.collect()

        if len(X_chunk_list) > 0:
            X_chunk = np.concatenate(X_chunk_list, axis=0).astype(np.float32)
            Y_chunk = np.concatenate(Y_chunk_list, axis=0).astype(np.float32)
            META_chunk = pd.concat(META_chunk_list, ignore_index=True)

            X_chunks.append(X_chunk)
            Y_chunks.append(Y_chunk)
            META_chunks.append(META_chunk)

            del X_chunk_list, Y_chunk_list, META_chunk_list, X_chunk, Y_chunk, META_chunk
            gc.collect()

    if len(X_chunks) == 0:
        return None, None, None

    X_all = np.concatenate(X_chunks, axis=0).astype(np.float32)
    Y_all = np.concatenate(Y_chunks, axis=0).astype(np.float32)
    META_all = pd.concat(META_chunks, ignore_index=True)

    assert len(X_all) == len(Y_all) == len(META_all)

    return X_all, Y_all, META_all
class PatchTSTRegression(nn.Module):
    def __init__(
        self,
        input_size=14,
        patch_len=32,
        stride=16,
        d_model=128,
        n_heads=4,
        n_layers=3,
        dropout=0.1
    ):
        super().__init__()

        self.patch_len = patch_len
        self.stride = stride

        patch_dim = patch_len * input_size
        self.patch_embed = nn.Linear(patch_dim, d_model)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=4 * d_model,
            dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model),
            nn.ReLU(),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        B, T, F = x.shape

        patches = []
        for i in range(0, T - self.patch_len + 1, self.stride):
            p = x[:, i:i+self.patch_len, :].reshape(B, -1)
            patches.append(p)

        patches = torch.stack(patches, dim=1)
        h = self.patch_embed(patches)
        z = self.transformer(h)

        z_last = z[:, -1, :]
        out = self.head(z_last).squeeze(-1)
        return out
def train_one_fold(X_train, Y_train, device, num_epochs=10, batch_size=512, lr=5e-4):
    train_ds = TensorDataset(
        torch.from_numpy(X_train).contiguous(),
        torch.from_numpy(Y_train).contiguous()
    )

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=0,
        pin_memory=True
    )

    model = PatchTSTRegression(input_size=len(FEATURE_COLS)).to(device)
    criterion = nn.SmoothL1Loss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    for epoch in range(num_epochs):
        model.train()
        train_loss_sum = 0.0
        train_count = 0

        pbar = tqdm(train_loader, desc=f"Train fold epoch {epoch+1}/{num_epochs}")

        for bx, by in pbar:
            bx = bx.to(device, non_blocking=True)
            by = by.to(device, non_blocking=True)

            pred = model(bx)
            loss = criterion(pred, by)

            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            train_loss_sum += loss.item() * bx.size(0)
            train_count += bx.size(0)

            pbar.set_postfix(batch_loss=float(loss))

        train_loss = train_loss_sum / max(train_count, 1)
        print(f"[Epoch {epoch+1}] train_loss={train_loss:.6f}")

    return model
def predict_with_meta(model, X_test, Y_test, META_test, device, batch_size=512):
    test_ds = TensorDataset(
        torch.from_numpy(X_test).contiguous(),
        torch.from_numpy(Y_test).contiguous()
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=True
    )

    preds = []
    trues = []

    model.eval()
    with torch.no_grad():
        for bx, by in test_loader:
            bx = bx.to(device, non_blocking=True)
            pred_batch = model(bx).cpu()
            preds.append(pred_batch)
            trues.append(by)

    pred = torch.cat(preds).numpy()
    true = torch.cat(trues).numpy()

    pred_df = META_test.copy().reset_index(drop=True)
    pred_df["pred_log_return"] = pred
    pred_df["true_log_return"] = true
    pred_df["pred_simple_return"] = np.expm1(pred_df["pred_log_return"])
    pred_df["true_simple_return"] = np.expm1(pred_df["true_log_return"])

    pred_df["hit_ge_10pct"] = (pred_df["true_simple_return"] >= 0.10).astype(int)
    pred_df["hit_ge_15pct"] = (pred_df["true_simple_return"] >= 0.15).astype(int)
    pred_df["hit_ge_20pct"] = (pred_df["true_simple_return"] >= 0.20).astype(int)

    return pred_df

def build_daily_watchlist(pred_df, top_n=10):
    df = pred_df.copy()
    df = df.dropna(subset=["anchor_date"]).copy()
    df["anchor_date"] = pd.to_datetime(df["anchor_date"])

    # 同一天同一股票，取最高分
    df = (
        df.groupby(["anchor_date", "stock_code"], as_index=False)
          .agg({
              "pred_log_return": "max",
              "pred_simple_return": "max",
              "true_log_return": "max",
              "true_simple_return": "max",
              "hit_ge_10pct": "max",
              "hit_ge_15pct": "max",
              "hit_ge_20pct": "max",
          })
    )

    daily_top = (
        df.sort_values(["anchor_date", "pred_log_return"], ascending=[True, False])
          .groupby("anchor_date", group_keys=False)
          .head(top_n)
          .reset_index(drop=True)
    )

    return daily_top
def build_weekly_watchlist(pred_df, top_n=10):
    df = pred_df.copy()
    df = df.dropna(subset=["anchor_date"]).copy()
    df["anchor_date"] = pd.to_datetime(df["anchor_date"])
    df["week"] = df["anchor_date"].dt.to_period("W").astype(str)

    df = (
        df.groupby(["week", "stock_code"], as_index=False)
          .agg({
              "anchor_date": "max",
              "pred_log_return": "max",
              "pred_simple_return": "max",
              "true_log_return": "max",
              "true_simple_return": "max",
              "hit_ge_10pct": "max",
              "hit_ge_15pct": "max",
              "hit_ge_20pct": "max",
          })
    )

    weekly_top = (
        df.sort_values(["week", "pred_log_return"], ascending=[True, False])
          .groupby("week", group_keys=False)
          .head(top_n)
          .reset_index(drop=True)
    )

    return weekly_top
def summarize_watchlist(watchlist_df, name="watchlist"):
    df = watchlist_df.copy()

    return pd.Series({
        "name": name,
        "n_rows": len(df),
        "n_unique_stocks": df["stock_code"].nunique() if "stock_code" in df.columns else np.nan,
        "avg_true_log_return": float(df["true_log_return"].mean()),
        "median_true_log_return": float(df["true_log_return"].median()),
        "avg_true_simple_return": float(df["true_simple_return"].mean()),
        "median_true_simple_return": float(df["true_simple_return"].median()),
        "hit_rate_ge_10pct": float(df["hit_ge_10pct"].mean()),
        "hit_rate_ge_15pct": float(df["hit_ge_15pct"].mean()),
        "hit_rate_ge_20pct": float(df["hit_ge_20pct"].mean()),
        "avg_pred_log_return": float(df["pred_log_return"].mean()),
    })

def walk_forward_backtest(
    parquet_files,
    feature_cols,
    time_step=126,
    future_days=60,
    window_stride=5,
    use_high_for_max=False,
    train_years=5,
    test_months=6,
    num_epochs=10,
    batch_size=512,
    lr=5e-4,
    device="cuda"
):
    """
    簡化版 walk-forward：
      - 每個 fold 用 train_end_date 前 train_years 的資料訓練
      - 預測接下來 test_months 的樣本
    """

    # 先掃所有可用日期範圍
    all_dates = []

    for pq in parquet_files:
        df = safe_read_parquet_features(pq, feature_cols, require_date=True)
        if df is None or df.empty:
            continue
        all_dates.append(df["Date"].min())
        all_dates.append(df["Date"].max())

    all_dates = pd.to_datetime(pd.Series(all_dates)).dropna()
    global_start = all_dates.min()
    global_end = all_dates.max()

    print("Global date range:", global_start.date(), "->", global_end.date())

    # 建立 fold 邊界
    # 例如 train 5 年，測 6 個月
    fold_starts = pd.date_range(
        start=global_start + pd.DateOffset(years=train_years),
        end=global_end - pd.DateOffset(months=test_months + 2),
        freq=f"{test_months}MS"
    )

    all_pred_df = []
    all_watchlist_summary = []

    for fold_i, train_end_date in enumerate(fold_starts, start=1):
        test_start_date = train_end_date + pd.Timedelta(days=1)
        test_end_date = train_end_date + pd.DateOffset(months=test_months)

        print("\n" + "="*80)
        print(f"[Fold {fold_i}] train_end={train_end_date.date()}  test={test_start_date.date()} -> {test_end_date.date()}")

        # 1) fit scaler using history up to train_end
        scaler = fit_global_scaler_until_date(parquet_files, feature_cols, train_end_date)

        # 2) build train set
        X_train_fold, Y_train_fold, META_train_fold = build_dataset_for_period(
            parquet_files=parquet_files,
            scaler=scaler,
            feature_cols=feature_cols,
            train_end_date=train_end_date,
            mode="train",
            time_step=time_step,
            future_days=future_days,
            window_stride=window_stride,
            use_high_for_max=use_high_for_max,
            chunk_size=20
        )

        # 3) build test set
        X_test_fold, Y_test_fold, META_test_fold = build_dataset_for_period(
            parquet_files=parquet_files,
            scaler=scaler,
            feature_cols=feature_cols,
            test_start_date=test_start_date,
            test_end_date=test_end_date,
            mode="test",
            time_step=time_step,
            future_days=future_days,
            window_stride=window_stride,
            use_high_for_max=use_high_for_max,
            chunk_size=20
        )

        if X_train_fold is None or X_test_fold is None:
            print("[SKIP fold] empty train or test")
            continue

        print("Train fold shapes:", X_train_fold.shape, Y_train_fold.shape, META_train_fold.shape)
        print("Test fold shapes :", X_test_fold.shape, Y_test_fold.shape, META_test_fold.shape)

        # 4) train model
        model = train_one_fold(
            X_train=X_train_fold,
            Y_train=Y_train_fold,
            device=device,
            num_epochs=num_epochs,
            batch_size=batch_size,
            lr=lr
        )

        # 5) predict test
        pred_df_fold = predict_with_meta(
            model=model,
            X_test=X_test_fold,
            Y_test=Y_test_fold,
            META_test=META_test_fold,
            device=device,
            batch_size=batch_size
        )

        pred_df_fold["fold_id"] = fold_i
        pred_df_fold["train_end_date"] = pd.Timestamp(train_end_date)
        pred_df_fold["test_start_date"] = pd.Timestamp(test_start_date)
        pred_df_fold["test_end_date"] = pd.Timestamp(test_end_date)

        # 6) build watchlists
        daily_top10 = build_daily_watchlist(pred_df_fold, top_n=10)
        weekly_top10 = build_weekly_watchlist(pred_df_fold, top_n=10)

        s_daily = summarize_watchlist(daily_top10, name=f"fold{fold_i}_daily_top10")
        s_weekly = summarize_watchlist(weekly_top10, name=f"fold{fold_i}_weekly_top10")

        s_daily["fold_id"] = fold_i
        s_weekly["fold_id"] = fold_i

        all_pred_df.append(pred_df_fold)
        all_watchlist_summary.append(s_daily)
        all_watchlist_summary.append(s_weekly)

        # 釋放 fold 記憶體
        del scaler, X_train_fold, Y_train_fold, META_train_fold, X_test_fold, Y_test_fold, META_test_fold, model, pred_df_fold, daily_top10, weekly_top10
        gc.collect()
        torch.cuda.empty_cache()

    if len(all_pred_df) == 0:
        raise ValueError("No valid walk-forward folds produced")

    pred_df_all = pd.concat(all_pred_df, ignore_index=True)
    watchlist_summary_df = pd.DataFrame(all_watchlist_summary)

    return pred_df_all, watchlist_summary_df
pred_df_wf, watchlist_summary_wf = walk_forward_backtest(
    parquet_files=parquet_files,
    feature_cols=FEATURE_COLS,
    time_step=TIME_STEP,
    future_days=FUTURE_DAYS,
    window_stride=WINDOW_STRIDE,
    use_high_for_max=USE_HIGH_FOR_MAX,
    train_years=5,
    test_months=6,
    num_epochs=10,
    batch_size=BATCH_SIZE,
    lr=LR,
    device=DEVICE
)

print("\n=== Walk-forward Watchlist Summary ===")
print(watchlist_summary_wf.to_string(index=False))

overall_summary = (
    watchlist_summary_wf
    .groupby(watchlist_summary_wf["name"].str.contains("weekly").map({True: "weekly", False: "daily"}))
    .agg({
        "avg_true_simple_return": "mean",
        "median_true_simple_return": "mean",
        "hit_rate_ge_10pct": "mean",
        "hit_rate_ge_15pct": "mean",
        "hit_rate_ge_20pct": "mean",
        "avg_pred_log_return": "mean",
    })
)

print("\n=== Overall average by watchlist type ===")
print(overall_summary)
```
